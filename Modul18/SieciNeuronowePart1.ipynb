# Podstawowe pojecia

- ML – skrót pojęcia Machine Learning

- DL – skrót pojęcia Deep Learning

- model – sieć neuronowa używana do rozwiązania konkretnego przykładu

- sieć neuronowa – inaczej zwana NN, Neural Network, podstawowy budulec algorytmów Deep Learningowych, który na bazie dostarczonych wejść, uczy się np. klasyfikować obiekty, wyznaczać, gdzie dany obiekt się znajduje itp.

- warstwa wejściowa – input layer, pierwsza warstwa sieci neuronowej. To do niej docierają dane wejściowe, które są następnie propagowane dalej

- warstwa wyjściowa – output layer, ostatnia warstwa sieci neuronowej, która dostarcza odpowiednie wyjście, np. wartości prawdopodobieństwa poszczególnych klas (czy dany obiekt to kot, pies czy czołg)

- funkcja aktywacyjna – activation function – to dzięki niej nasza sieć nie jest modelem liniowym, który tylko przekazuje dane wejściowe i mnoży je przez odpowiednie współczynniki. Wprowadza nieliniowość do naszego modelu. Bez nieliniowości nie ma sieci neuronowych

- optymalizator – optimizer – obiekt służący do nauki naszej sieci neuronowej poprzez odpowiednie modyfikowanie wag poszczególnych elementów sieci

- ML framework – określenie na bibliotekę używaną do stworzenia/nauki sieci neuronowej

- funkcja błędu – loss function, loss – funkcja mówiąca nam, jak bardzo nasze wartości są niezgodne z oczekiwanymi wartościami. Na bazie jej obliczeń odbywa się korekcja wag przez optimizer

- forward propagation – przekazywanie danych od wejścia do wyjścia sieci

- back propagation – propagacja wsteczna, główny mechanizm szkolenia sieci neuronowej. Na bazie wyników loss function przy pomocy odpowiednich metod matematycznych wagi naszej sieci są modyfikowane, przez co osiągany jest efekt uczenia się. Szczegóły matematyczne tej tematyki są poza zakresem tego kursu.

- Gradient Descent – metoda optymalizacji sieci neuronowych używana przez optimizer.

następnie dane przekazywane są do kolejnej warstwy za pomocą połączeń każdy do każdego, jak pokazano na poniższym rysunku.

Jest tutaj kilka rzeczy, które należy zaznaczyć:

każde połączenie między neuronami posiada swój własny zestaw wewnętrznych wag (na rysunku zaznaczone są tylko wagi w00, w01 itd. pomiędzy warstwą wejściową oraz ukrytą dla pierwszego neuronu wyjściowego, lecz należy mieć na uwadze, iż wszystkie kreski między wszystkimi neuronami mają takie wagi).

Czym jest bias? Bias jest dodatkowym składnikiem dodawanym do wartości wagi naszej warstwy, aby wprowadzić pewnego rodzaju "zakłócenia", które wprowadza opóźnienie w aktywacji naszego neuronu. Do czego jest to potrzebne? Aby wprowadzić jeszcze większą nieliniowość do naszego systemu, oraz aby dodać pewną stałą, która da nam większą pewność, że nasze wagi nie będą zerami.

x_size – jest to ilość neuronów warstwy poprzedniej – h_size - jest to ilość neuronów warstwy aktualnej

w00[0] = w00+b

gdzie w00[0] (macierz łącząca warstwę wejściową z kolejną) jest macierzą o rozmiarze h_size x x_size, b jest macierzą o rozmiarze h_size x 1.

dane wpływające do danego neuronu są sumą wszystkich wejść od neuronów z poprzedniej warstwy dla danego neuronu. Przykładowe wejście do neuronu y1 pokazuje poniższy obrazek.

Teraz pora na troszkę matematyki (wzór (1)):

$$y1=wyjście_h0w01+wyjście_h1w11+wyjście_h2w21+wyjście_h3w31$$

Jak powiedzieliśmy, sumujemy wszystko, co wchodzi do danego neuronu i voilà – mamy wartość wejściową dla naszego pojedynczego neuronu.

- Suma wejść danego neuronu jest znowu mnożona przez wagi i przekazywana do kolejnej warstwy.

# Funkcje aktywacyjne

Funkcja aktywacyjna przekształca wartość obliczoną przez neuron:

$$z = w_1x_1 + w_2x_2 + \ldots + w_nx_n + b$$

w jego wyjście:

$$a = f(z)$$

Bez nieliniowych funkcji aktywacyjnych wiele kolejnych warstw `Dense` nadal zachowywałoby się jak pojedyncza transformacja liniowa. Nieliniowość pozwala sieci uczyć się bardziej złożonych zależności, których nie da się opisać prostą lub hiperpłaszczyzną.

## Najpopularniejsze funkcje aktywacyjne

````python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

# Zakres wartości przekazywanych do funkcji aktywacyjnych.
x = np.linspace(-6, 6, 500, dtype=np.float32)
x_tensor = tf.constant(x)

activation_values = {
    "Linear": x,
    "Sigmoid": tf.nn.sigmoid(x_tensor).numpy(),
    "Tanh": tf.nn.tanh(x_tensor).numpy(),
    "ReLU": tf.nn.relu(x_tensor).numpy(),
    "Leaky ReLU": tf.nn.leaky_relu(x_tensor, alpha=0.1).numpy(),
    "ELU": tf.nn.elu(x_tensor).numpy()
}

fig, axes = plt.subplots(2, 3, figsize=(15, 8))

for ax, (name, values) in zip(axes.flat, activation_values.items()):
    ax.plot(x, values, linewidth=2)
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_title(name)
    ax.set_xlabel("z")
    ax.set_ylabel("f(z)")
    ax.grid(alpha=0.25)

plt.suptitle("Porównanie funkcji aktywacyjnych", fontsize=16)
plt.tight_layout()
plt.show()
````

### Interpretacja

- **Linear** — nie wprowadza nieliniowości. Najczęściej jest używana w warstwie wyjściowej regresji, gdy model może zwracać dowolną liczbę rzeczywistą.
- **Sigmoid** — zwraca wartości od 0 do 1, dlatego dobrze opisuje prawdopodobieństwo klasy dodatniej w klasyfikacji binarnej. Dla dużych wartości bezwzględnych gradient jest bardzo mały, więc sigmoid zwykle nie jest stosowany w głębokich warstwach ukrytych.
- **Tanh** — zwraca wartości od -1 do 1 i jest wycentrowana wokół zera. Podobnie jak sigmoid może cierpieć na problem zanikającego gradientu.
- **ReLU** — zwraca 0 dla wartości ujemnych i wartość wejściową dla dodatnich. Jest szybka i stanowi najczęstszy wybór dla warstw ukrytych. Neurony mogą jednak przestać się aktualizować, jeśli stale otrzymują wartości ujemne.
- **Leaky ReLU** — dla wartości ujemnych pozostawia niewielkie nachylenie. Ogranicza w ten sposób problem „martwych neuronów” ReLU.
- **ELU** — dla wartości ujemnych zwraca gładkie wartości zbliżające się do -1. Może poprawić przepływ gradientu, ale jest obliczeniowo bardziej złożona od ReLU.

## Softmax w klasyfikacji wieloklasowej

````python
# Logity dla jednej obserwacji należącej do jednej z trzech klas.
logits = tf.constant([[2.0, 1.0, 0.1]])
probabilities = tf.nn.softmax(logits, axis=1)

print("Logity:", logits.numpy())
print("Prawdopodobieństwa:", probabilities.numpy().round(4))
print("Suma prawdopodobieństw:", probabilities.numpy().sum(axis=1))
````

`Softmax` zamienia zestaw logitów na rozkład prawdopodobieństwa. Każda wartość mieści się w przedziale od 0 do 1, a ich suma dla jednej obserwacji wynosi 1. Funkcję tę stosujemy w warstwie wyjściowej klasyfikacji wieloklasowej, gdy każda obserwacja należy dokładnie do jednej klasy.

## Dobór funkcji aktywacyjnej do rodzaju problemu

````python
from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Dense

# Wspólny przykład danych wejściowych zawierających 10 cech.
number_of_features = 10

# 1. Klasyfikacja binarna — jedna wartość od 0 do 1.
binary_model = Sequential([
    Input(shape=(number_of_features,)),
    Dense(32, activation="relu"),
    Dense(1, activation="sigmoid")
])
binary_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# 2. Klasyfikacja wieloklasowa — rozkład prawdopodobieństwa klas.
number_of_classes = 3
multiclass_model = Sequential([
    Input(shape=(number_of_features,)),
    Dense(32, activation="relu"),
    Dense(number_of_classes, activation="softmax")
])
multiclass_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# 3. Regresja — wyjście liniowe może przyjmować dowolną wartość.
regression_model = Sequential([
    Input(shape=(number_of_features,)),
    Dense(32, activation="relu"),
    Dense(1, activation="linear")
])
regression_model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

print("Utworzono trzy poprawnie skonfigurowane modele demonstracyjne.")
````

### Najważniejsza zasada

Funkcja aktywacyjna ostatniej warstwy musi pasować do zadania i funkcji straty:

| Rodzaj problemu | Warstwa wyjściowa | Typowa funkcja straty |
|---|---|---|
| Regresja | `Dense(1, activation="linear")` | MSE lub MAE |
| Klasyfikacja binarna | `Dense(1, activation="sigmoid")` | Binary Crossentropy |
| Klasyfikacja wieloklasowa | `Dense(n_classes, activation="softmax")` | Categorical lub Sparse Categorical Crossentropy |

Warstwy ukryte najczęściej wykorzystują ReLU lub jeden z jej wariantów. Sigmoid i softmax stosujemy przede wszystkim w warstwie wyjściowej odpowiedniego problemu klasyfikacyjnego.

````python
# Importujemy bibliotekę TensorFlow, która służy do budowania i trenowania sieci neuronowych.
import tensorflow as tf

# Importujemy model Sequential, w którym warstwy są wykonywane kolejno.
from tensorflow.keras.models import Sequential

# Importujemy warstwę Input, która definiuje kształt danych wejściowych.
from tensorflow.keras.layers import Input

# Importujemy warstwę Conv2D, która wykrywa cechy na obrazach.
from tensorflow.keras.layers import Conv2D

# Importujemy warstwę MaxPooling2D, która zmniejsza rozmiar map cech.
from tensorflow.keras.layers import MaxPooling2D

# Importujemy warstwę Flatten, która zamienia mapy cech na wektor.
from tensorflow.keras.layers import Flatten

# Importujemy warstwę Dense, czyli warstwę w pełni połączoną.
from tensorflow.keras.layers import Dense


# Ustalamy wysokość każdego obrazu wejściowego na 150 pikseli.
IMG_HEIGHT = 150

# Ustalamy szerokość każdego obrazu wejściowego na 150 pikseli.
IMG_WIDTH = 150


# Tworzymy model sekwencyjny, w którym dane przechodzą przez warstwy po kolei.
model = Sequential([

    # Definiujemy wejście modelu jako obraz o wysokości 150, szerokości 150 i trzech kanałach RGB.
    Input(
        shape=(IMG_HEIGHT, IMG_WIDTH, 3)  # kształt pojedynczego obrazu wejściowego
    ),

    # Tworzymy pierwszą warstwę konwolucyjną.
    Conv2D(
        filters=16,           # liczba filtrów wykrywających różne cechy obrazu
        kernel_size=(3, 3),   # rozmiar każdego filtra wynosi 3x3 piksele
        padding='same',       # zachowujemy wysokość i szerokość mapy cech
        activation='relu'     # zerujemy wartości ujemne i zachowujemy dodatnie
    ),

    # Zmniejszamy szerokość i wysokość map cech za pomocą okna 2x2.
    MaxPooling2D(
        pool_size=(2, 2)      # z każdego obszaru 2x2 wybieramy największą wartość
    ),

    # Tworzymy drugą warstwę konwolucyjną.
    Conv2D(
        filters=32,           # zwiększamy liczbę filtrów do 32
        kernel_size=(3, 3),   # każdy filtr nadal ma rozmiar 3x3
        padding='same',       # zachowujemy przestrzenny rozmiar map cech
        activation='relu'     # stosujemy funkcję aktywacji ReLU
    ),

    # Ponownie zmniejszamy szerokość i wysokość map cech.
    MaxPooling2D(
        pool_size=(2, 2)      # wybieramy maksimum z każdego obszaru 2x2
    ),

    # Tworzymy trzecią warstwę konwolucyjną.
    Conv2D(
        filters=64,           # wykorzystujemy 64 filtry do wykrywania bardziej złożonych cech
        kernel_size=(3, 3),   # każdy filtr analizuje fragment obrazu 3x3
        padding='same',       # zachowujemy wysokość i szerokość map cech
        activation='relu'     # wprowadzamy nieliniowość za pomocą ReLU
    ),

    # Po raz trzeci zmniejszamy rozmiar map cech.
    MaxPooling2D(
        pool_size=(2, 2)      # redukujemy rozmiar danych za pomocą okna 2x2
    ),

    # Zamieniamy trójwymiarowe mapy cech na jeden długi wektor liczb.
    Flatten(),

    # Tworzymy warstwę w pełni połączoną.
    Dense(
        units=512,            # warstwa zawiera 512 neuronów
        activation='relu'     # neurony wykorzystują funkcję aktywacji ReLU
    ),

    # Tworzymy warstwę wyjściową dla klasyfikacji binarnej.
    Dense(
        units=1,              # jeden neuron zwraca wynik dla klasy dodatniej
        activation='sigmoid'  # zamieniamy wynik na prawdopodobieństwo od 0 do 1
    )
])


# Konfigurujemy sposób trenowania modelu.
model.compile(

    # Używamy optymalizatora Adam do aktualizowania wag sieci.
    optimizer='adam',

    # Używamy funkcji straty przeznaczonej do klasyfikacji binarnej.
    loss=tf.keras.losses.BinaryCrossentropy(
        from_logits=False     # False, ponieważ sigmoid zwraca już prawdopodobieństwo
    ),

    # Podczas trenowania monitorujemy dokładność klasyfikacji.
    metrics=[
        'accuracy'             # procent prawidłowo sklasyfikowanych obrazów
    ]
)


# Wyświetlamy nazwy warstw, kształty ich wyjść i liczbę parametrów.
model.summary()
````

 MaxPooling2D - Podsumowując: pooling stosujemy, aby zmniejszyć koszty obliczeń, zwiększyć pole widzenia i ograniczyć przeuczenie. Nie trzeba go jednak dodawać po każdej warstwie konwolucyjnej — częstotliwość zależy od architektury i rodzaju analizowanych obrazów.

# Sigmoid

Ta funkcja jest odpowiednikiem regresji logistycznej, jej wyjście powinno być dla Ciebie znane – kształtem przypomina odwróconą literę S.

f(x) = 1/ 1+e (do -x)

Powodem jej wykorzystywania jest uzyskanie wartości z zakresu od -1 do 1. Wiemy już, że idealnie pasuje do klasyfikacji binarnej, ponieważ uzyskany wynik jest tożsamy z prawdopodobieństwem. Funkcja ta, w przeciwieństwa do funkcji progowej, jest całkowalna. W związku z tym stosować można metodę gradientu prostego (taka metoda do wyboru wag) i trenować sieć w celu optymalizacji. Jest tak, ponieważ funkcja sigmoidalna w każdym punkcie ma niezerową pochodną. Funkcja sigmoidalna do pewnego czasu była najpopularniejszą, wyparła ją ReLU, ponieważ w przeciwieństwie do funkcji sigmoidalnej nia ma wady zanikającego gradientu. Funkcja sigmoidalna przekształca dane wejściowe w wyjście z zakresu od 0 do 1. Gdy iloczyn danych wejściowych oraz wag mają wartości zmierzające do -∞ lub +∞, pochodna jest bliska 0. W przypadku wielu warstw gradient może być tak mały, że proces uczenia staje się nieefektywny, ponieważ nie aktualizuje się wag w warstwach początkowych.

Sigmoid przekształca dowolną liczbę na wartość od 0 do 1:
\[
\sigma(x)=\frac{1}{1+e^{-x}}
- Najważniejsze zastosowanie to warstwa wyjściowa w klasyfikacji binarnej, np.:
    - spam lub nie spam,
   - kot lub pies,
    -klient kupi lub nie kupi,
     -choroba występuje lub nie występuje.
Dense(1, activation='sigmoid')

Wynik można interpretować jako prawdopodobieństwo klasy pozytywnej:
0,10 → około 10%
0,50 → około 50%
0,90 → około 90%
Najczęściej stosuje się próg 0.5:
predicted_class = 1 if probability >= 0.5 else 0


Sigmoid rzadko stosuje się w warstwach ukrytych, ponieważ dla bardzo dużych wartości dodatnich i ujemnych jej gradient zanika, co spowalnia uczenie sieci.

# Tanh – Funkcja tangensa hiperbolicznego

Podobnie jak funkcja sigmoidalna przypomina kształtem literę S. Ona również jest ciągła i różniczkowalna.

Wartość wyjściowa jest lepiej wyśrodkowana niż funkcja sigmoidalna, ponieważ jest ona wyśrodkowana względem 0, a jej zakres jest od -1 do 1. W praktyce optymalizacja jest łatwiejsza niż dla funkcji sigmoidalnej, jednak nadal występuje problem zanikającego gradientu.

Najważniejsze cechy:
- wartości ujemne są przekształcane na wyniki bliskie -1,
- dla x = 0 wynik wynosi 0,
- wartości dodatnie dają wyniki bliskie 1,
- funkcja jest wycentrowana wokół zera.
- 
Tanh stosuje się głównie:
- w warstwach ukrytych prostszych sieci,
- w sieciach rekurencyjnych, np. RNN i LSTM,
- gdy istotne są zarówno wartości dodatnie, jak i ujemne.
  
Dense(32, activation='tanh')

Tanh może cierpieć na problem zanikającego gradientu dla dużych wartości dodatnich i ujemnych. Dlatego w wielu współczesnych sieciach warstwy ukryte częściej wykorzystują ReLU.

# Funkcja ReLU - Rectified Linear Unit

ReLU zwraca 0 dla wartości ujemnych, a wartości dodatnie pozostawia bez zmian:

ReLU(x)=\max(0,x)


Najważniejsze cechy:
- jest prosta i szybka obliczeniowo,
- wprowadza nieliniowość do modelu,
- pomaga ograniczyć problem zanikającego gradientu,
- jest najczęściej stosowana w warstwach ukrytych.
    
Najważniejsze zastosowania:
- sieci konwolucyjne CNN,
- klasyczne sieci neuronowe,
- przetwarzanie obrazów,
- głębokie warstwy ukryte.

    
Dense(32, activation='relu')
lub:
Conv2D(32, 3, activation='relu')

    
Wadą ReLU jest problem „martwych neuronów”. Jeśli neuron stale otrzymuje wartości ujemne, zawsze zwraca 0 i może przestać się uczyć. W takim przypadku można zastosować np. Leaky ReLU.


Jest to kombinacja funkcji progowej oraz liniowej. Obecnie najpopularniejsza funkcja aktywacji, jeśli nie wiesz którą wykorzystać – spróbuj ReLU, zazwyczaj działa najlepiej i bezproblemowo. Matematycznie oraz intuicyjnie jest to prosta funkcja aktywacji, a przy tym jest skuteczna. Co najważniejsze nie występuje tu problem zanikającego gradientu. Nie jest to jednak idealna funkcja, ponieważ dla wartości z bloku sumującego mniejszych niż 0, zawsze będzie 0, nieważne czy to -1 czy -1000. 

# Funkcja Softmax

Klasyfikacja nie musi być binarna, może być wieloklasowa. Dla tego problemu stosujemy funkcję Softmax. Na wejściu otrzymujemy wektor o długości i równej liczbie klas. Na wyjściu otrzymujemy wektor prawdopodobieństw przynależności do klasy j, sumujący się do 1.

Softmax zamienia zestaw wyników modelu na prawdopodobieństwa poszczególnych klas.
    
Najważniejsze cechy:
- każda wartość mieści się w przedziale od 0 do 1,
- suma wszystkich wyników wynosi 1,
- największe prawdopodobieństwo wskazuje przewidywaną klasę.
    
Przykład wyniku:
kot       → 0,80
pies      → 0,15
samochód  → 0,05
Model wybierze klasę „kot”.
    
Najważniejsze zastosowanie to warstwa wyjściowa w klasyfikacji wieloklasowej, gdy obserwacja należy dokładnie do jednej klasy:
Dense(3, activation='softmax')
Softmax najczęściej łączy się z funkcją straty:
loss='sparse_categorical_crossentropy'
    
Nie stosuje się go zwykle w warstwach ukrytych. W klasyfikacji binarnej częściej używa się jednego neuronu z funkcją sigmoid.

# Optimizer

Kolejną rzeczą używaną podczas szkolenia sieci neuronowych jest byt zwany optimizer.

Jest on odpowiedzialny za wyliczanie gradientów, czyli kierunku oraz wartości zmian wag naszej sieci (każdej warstwy sieci oraz każdego neuronu w tej warstwie), bazując na aktualnych wynikach naszej sieci.

Stosuje on metodę tzw. Gradient Descent. Polega ona na wykonywaniu kroków w kierunku oraz o wartości wyznaczonych przez wartość pochodnej danej zmiennej, pomnożonej przez współczynnik tzw. learning rate. Powoduje to systematyczne zbliżanie się do minimum funkcji, którą optymalizujemy.

Optimizer odpowiada za aktualizowanie wag i biasów sieci neuronowej podczas trenowania. 
Jego celem jest takie zmienianie parametrów modelu, aby zmniejszać wartość funkcji straty.
    
Proces wygląda następująco:
- model wykonuje predykcję,
- funkcja straty oblicza błąd,
- backpropagation oblicza gradienty,
- optimizer aktualizuje wagi modelu.
    
Najważniejszym parametrem optymalizatora jest learning_rate, czyli tempo uczenia:
- zbyt duże — model może omijać dobre rozwiązanie,
- zbyt małe — model uczy się bardzo wolno.
    
Najczęściej stosowane optymalizatory:
- Adam — uniwersalny i często dobry wybór początkowy,
- SGD — prosty, ale może wymagać dokładniejszego strojenia,
- RMSprop — często wykorzystywany w modelach sekwencyjnych, np. RNN.
    
Przykład:
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
Optymalizator z własnym tempem uczenia:
optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.001
)

model.compile(
    optimizer=optimizer,
    loss='binary_crossentropy',
    metrics=['accuracy']
)

Optimizer nie określa, co model ma przewidywać.
Odpowiada jedynie za sposób, w jaki model uczy się na podstawie obliczonego błędu.

W pierwszym kroku losujemy początkowe wagi, następnie zgodnie z współczynnikiem uczenia α minimalizujemy funkcję kosztu (błędy) i dobieramy lepsze wagi. Jeśli dobrana zostanie zbyt niska wartość α, to algorytm będzie musiał wykonać wiele iteracji do momentu uzyskania optymalnych wartości wag. W wyniku tych operacji uczenie będzie trwało długo. Innym przypadkiem jest dobranie zbyt wysokiej wartości współczynnika uczenia α. Algorytm może „przeskoczyć” minimum, zwiększać wartość straty i stać się niezdolny do znalezienia optymalnych wartości wag. Miej na uwadze, że tutaj dobieramy jedynie jeden współczynnik, zazwyczaj sieci neuronowe pracują na większej liczbie zmiennych, wtedy mamy minimum globalne oraz wiele minimum lokalnych. Chcemy tak nauczyć model, aby znalazł minimum globalne.

Bez zagłębiania się w szczegóły, zapoznajmy się z podstawowymi optimizerami stosowanymi w ML:

- SGD – Stochastic Gradient Descent, najprostszy z optimizerów wykonujący równanie opisane powyżej. Dobry jako domyślny, jednak są lepsze. Nadaje się do segmentacji obrazów, jednak wymaga użycia kilku bardziej zaawansowanych sztuczek, aby zadziałał. Stosowany m.in. w Yolov5, aktualnie w Top 2 modeli do segmentacji obrazów.

- Adam – Adaptive Momentum Optimizer, wykorzystuje adaptatywny learning rate, jest rozszerzeniem SGD, zazwyczaj jest od niego lepszy, powinien być używany jako domyślny.

- AdamW – Adam Weighted, wersja Adama na sterydach, lecz nie jest ona dołączona domyślnie w TensorFlow. Znajduje się w dodatkowej bibliotece tensorflow_addons.

- Ranger, jeden z lepszych optimizerów, lecz wymaga osobnego pobrania i zbudowania. Instrukcja, jak to zrobić znajduje się w załączonym linku.

# Learning_rate

Jest to jeden z najważniejszych hiperparametrów w Machine Learningu, gdyż określa, jak szybko nasz model będzie się uczył. Jest on trudny do ustawienia za pierwszym razem i manewrowanie nim wymaga troszkę wprawy, lecz należy trzymać się podstawowych zasad:

- jeśli nasz model się nie uczy (metryka nie wzrasta), należy zwiększyć learning_rate,

- jeśli metryka naszego zbioru walidacyjnego się nie zmniejsza, natomiast zbioru treningowego się zmniejsza, należy go zmniejszyć.

Oczywiście są to zasady, które sprawdzają się często, ale nie zawsze.

# Schedulers

Schedulery są metodami, które pozwalają nam zmieniać wartość learning_rate podczas nauki naszego modelu. Dlaczego właściwie zmieniać learning_rate podczas nauki?

Okazuje się, że bardzo często daje to o wiele lepsze wyniki niż stała wartość. Głównym powodem stosowania schedulerów jest próba uniknięcia lokalnego minimum. Celem naszej optymalizacji jest znalezienie minimum globalnego, czyli zestawu wag, które dadzą nam najmniejszą różnicę pomiędzy przewidywanymi oraz rzeczywistymi wartościami. Jednak podczas naszych poszukiwań może się zdarzyć, iż utkniemy w tzw. lokalnym minimum. Nie wyjdziemy z niego natomiast jeśli learning rate jest za duże. Zbadajmy, jaki jest tego powód, aby zrozumieć ideę, która się za tym kryje.

Uczenie się przez sieć można porównać do nauki nowych rzeczy przez człowieka:

- na początku, kiedy uczymy się np. programowania, nie posiadamy żadnej wiedzy na ten temat. W związku z tym każda nowa informacja powiększa nasz zasób wiedzy, słownictwa itp. szybko wchłaniamy dużo wiedzy, czytamy całe książki itd.

- Jednak wraz z czasem, uczymy się coraz mniej, ponieważ znamy już ogólny zarys i zaczynamy szukać informacji o specyficznych rzeczach, np. zamiast uczyć się wszystkiego o funkcjach w Pythonie, po jakimś czasie uczymy się, jak uczynić funkcje szybszymi. Nie przyswajamy więc tak wielkiej ilości informacji, jak na początku i nie poświęcamy tak wiele czasu na czytanie całych książek, szukamy konkretnych rozdziałów.

- W związku z tym przyswajamy mniej informacji, ale bardziej specyficznych, musimy skupić się na węższym zakresie danych.

Ze sztucznymi sieciami neuronowymi jest identycznie jak z tymi w naszym mózgu.

- Na początku potrzebujemy większego learning_rate, aby nauczyć się ogólnych informacji na temat zbioru wejściowego, poznać zasady rządzące jego rozkładem, pikselami itp., np. taki kształt to zwierzę, a taki to nie zwierzę.

- Z czasem, kiedy już poznaliśmy ogólne idee naszego zbioru wejściowego, powinniśmy skupić się na bardziej szczegółowych informacjach np. taki kolor i taki pyszczek to kot syberyjski, a taki kolor plus niebieskie oczy, plus "nie chodzi na czworaka", to człowiek.

- Te szczegółowe informacje łatwiej wyłapiemy, zmniejszając wartość learning_rate.

W TensorFlow istnieje twór zwany LearningRateScheduler, który właśnie temu służy.

Jest to klasa, której przekazujemy naszą własną funkcję - "ustawiacz" learning_rate.

Widać, że aby wykorzystać tę technikę musimy przygotować funkcję przyjmującą jako argumenty numer epoki i aktualny learning_rate a zwracającą nową jego wartość i opakowawszy ją w LearningRateScheduler przekazać jako callback metody fit, czyli do procesu nauki naszego modelu.

````python
model.fit(
    ...,
    callbacks=[
        tf.keras.callbacks.LearningRateScheduler(wlasna_funkcja)
    ]
)
````

Metody typu callback, przekazane do parametru callback, są wywoływane pod koniec nauki każdej epoki, więc zmiana learning_rate będzie odbywała się co epokę.

Nie zawsze istnieje konieczność pisania własnej funkcji. W TensorFlow znajduje się gotowy scheduler, który powoduje spadek learning_rate cosinusoidalnie – CosineDecay.

Schedulery mogą zmieniać wartość learning_rate na wiele sposobów, przykładowo:

- co n epok zmniejsz LR o połowę,
- zmniejszaj LR liniowo co epokę, aż do ustalonej wartości,
- zmniejszaj LR ze spadkiem cosinusoidalnym,
- utrzymuj LR równo przez n epok następnie zmniejszaj,
- utrzymuj wartość startową co n epok, następnie przez m epok zmniejszaj, potem wróć do wartości startowej i powtórz.

Najprostszym sposobem, aby uzyskać zmianę learning_rate jest ustawienie parametru decay w optimizerach (jest to parametr klasy bazowej, więc trzeba go szukać tutaj).

adam_opt = keras.optimizers.Adam(lr=0.01, decay=1e-5)
Exponential scheduling ze stałą wartością learning_rate – zmiana wartości learning_rate wykładniczo:

def exponential_decay_fn(epoch):
	return 0.05 * 0.2**(epoch / 10)
Exponential scheduling z wartością zależną od aktualnego learning_rate:

def exponential_decay_fn(epoch, current_lr):
  return current_lr * 0.2**(epoch / 10)
Scheduler zależny od numeru epoki:

dla pierwszych epok ustaw duża wartość
dla epok do 20 ustaw mniejszą wartość
dla pozostałych ustaw małą wartość
def const_scheduler(epoch):
    if epoch < 10:
          return 0.05
    elif epoch < 20:
          return 0.005
    else:
          return 0.0005

# Scheduler — harmonogram learning rate

Scheduler automatycznie zmienia learning_rate podczas trenowania.
Najczęściej:

- na początku stosuje większy krok, aby model uczył się szybko,
- później zmniejsza krok, aby dokładniej dopasować wagi,
- pomaga ustabilizować uczenie pod koniec treningu,
- może pomóc, gdy val_loss przestaje się poprawiać.
- 
W aktualnym Kerasie optimizer przyjmuje learning_rate jako liczbę albo obiekt harmonogramu. Stary parametr decay nie występuje już w konstruktorze Adam; weight_decay to inny mechanizm i nie zastępuje schedulera. Dokumentacja Keras Adam

Który scheduler wybrać?
- LearningRateScheduler — gdy chcesz samodzielnie określić zmianę dla każdej epoki.
- ExponentialDecay — gdy chcesz płynnie zmniejszać learning rate według liczby partii.
- scheduler schodkowy — gdy z góry znasz momenty zmiany.
- ReduceLROnPlateau — gdy learning rate ma reagować na brak poprawy val_loss.
- 
Na początek najbardziej praktycznym rozwiązaniem jest często ReduceLROnPlateau, ponieważ reaguje na rzeczywisty przebieg trenowania, zamiast zakładać z góry, kiedy model przestanie się poprawiać.

Zanim jednak zaczniemy stosować schedulery, warto wypróbować jeszcze jeden przydatny callback dostępny w TensorFlow – ReduceLROnPlateau. Sprawdza on metrykę, którą mu podamy w parametrze var_loss (domyślnie sprawdza błąd zestawu walidacyjnego, więc zazwyczaj jest to OK). Jeżeli wykryje, że przez patience epok nie ma poprawy tej metryki, czyli nasz model nie uczy się lepiej generalizować na niewidzianych danych, zmniejsza learning_rate o współczynnik określony zmienną factor. Tak samo, jak w poprzednim przypadku, callback ten przekazuje się podczas szkolenia naszego modelu jako element listy w parametrze callbacks metody fit.

# Loss Functions – funkcje błędu

Kolejną istotną rzeczą podczas trenowania naszego modelu jest funkcja błędu, loss function. Odzwierciedla ona postęp nauki i jest minimalizowana w jej trakcie.

Istnieje wiele możliwych funkcji błędu, lecz najczęściej stosuje się tylko kilka. Podstawowe oraz wystarczające w większości przypadków loss functions znajdują się pod tym linkiem.

Przeglądając występujące tam funkcje widać, że wiele posiada w nazwie Crossentropy. Dzieje się tak, ponieważ Crossentropy jest jedną z najpowszechniejszych funkcji błędu oraz bazą do wielu innych. De facto większość z nich odnosi się do specyficznych przypadków (są one podzielone, aby uprościć matematykę, a przez to efektywnie przyspieszyć naukę naszego modelu). Przyjrzyjmy się, jakie występują warianty tej funkcji straty:

- CategoricalCrossentropy
- BinaryCrossentropy
- SparseCategoricalCrosentropy

Funkcja straty mierzy różnicę między prawidłowym wynikiem y_true a predykcją modelu y_pred.
    
Podczas trenowania:
- model wykonuje predykcję,
- funkcja straty oblicza błąd,
- backpropagation oblicza gradienty,
- optimizer zmienia wagi, aby zmniejszyć stratę.
    
Im mniejsza wartość loss, tym lepiej predykcje modelu pasują do danych treningowych.

# Binary Crossentropy


Stosowana w klasyfikacji binarnej, czyli przy dwóch możliwych klasach:
- spam lub ham,
- kot lub pies,
- prawda lub fałsz.
    
Etykiety mają postać:
y = [0, 1, 1, 0]

Warstwa wyjściowa zawiera jeden neuron z funkcją sigmoid:
Dense(1, activation='sigmoid')

model.compile(
    optimizer='adam',
    loss=tf.keras.losses.BinaryCrossentropy(
        from_logits=False
    ),
    metrics=['accuracy']
)
from_logits=False, ponieważ sigmoid zwraca już prawdopodobieństwo od 0 do 1.
    
Binary Crossentropy można również stosować w klasyfikacji wieloetykietowej, gdy jedna obserwacja może jednocześnie należeć do kilku klas.

# Categorical Crossentropy

Stosowana w klasyfikacji wieloklasowej, gdy każda obserwacja należy dokładnie do jednej klasy, a etykiety są zapisane metodą one-hot encoding.
    
Przykład trzech klas:
- kot  → [1, 0, 0]
- pies → [0, 1, 0]
- ptak → [0, 0, 1]
    
Warstwa wyjściowa musi mieć tyle neuronów, ile istnieje klas:
Dense(3, activation='softmax')

model.compile(
    optimizer='adam',
    loss=tf.keras.losses.CategoricalCrossentropy(
        from_logits=False
    ),
    metrics=['accuracy']
)
Softmax zwraca prawdopodobieństwo każdej klasy, a suma wyników wynosi 1.

# Sparse Categorical Crossentropy

Stosowana w tym samym rodzaju problemu co CategoricalCrossentropy, ale etykiety są zapisane jako pojedyncze liczby całkowite.
    
Przykład:
- kot  → 0
- pies → 1
- ptak → 2
    
y = [0, 2, 1, 0]
  
Nie trzeba wykonywać one-hot encoding.
    
Warstwa wyjściowa:
    
Dense(3, activation='softmax')

model.compile(
    optimizer='adam',
    loss=tf.keras.losses.SparseCategoricalCrossentropy(
        from_logits=False
    ),
    metrics=['accuracy']
)

Najważniejsza różnica
Funkcja straty	Rodzaj problemu	Przykładowa etykieta	Wyjście
BinaryCrossentropy	Dwie klasy	0 lub 1	1 neuron + sigmoid
CategoricalCrossentropy	Wiele klas	[0, 1, 0]	n neuronów + softmax
SparseCategoricalCrossentropy	Wiele klas	1	n neuronów + softmax

CategoricalCrossentropy i SparseCategoricalCrossentropy rozwiązują ten sam problem. Różnią się wyłącznie sposobem zapisania prawidłowych etykiet.
    

Sparse Categorical Crossentropy
Jeżeli y jest numerem prawidłowej klasy, równanie ma prostą postać:

L=-log(p_y)

gdzie:
-\(L\) — wartość błędu,
-\(y\) — numer prawidłowej klasy,
-\(p_y\) — prawdopodobieństwo przypisane przez model prawidłowej klasie.
    
Przykład:
Prawidłowa klasa: pies
Prawdopodobieństwa modelu:

kot  = 0,10
pies = 0,80
ptak = 0,10
Strata:

L=-log(0.8)approx 0.223

\]Jeżeli model przypisze prawidłowej klasie małe prawdopodobieństwo:
\[
L=-\log(0.1)\approx 2.303
\]Im większe prawdopodobieństwo prawidłowej klasy, tym mniejszy błąd.

Categorical Crossentropy
Dla etykiet zapisanych metodą one-hot encoding:

L=-\sum_{i=1}^{K} y_i\log(p_i)
\]gdzie:
\(K\) — liczba klas,
\(y_i\) — wartość prawdziwej etykiety dla klasy \(i\), czyli 0 albo 1,
\(p_i\) — prawdopodobieństwo klasy \(i\).
Ponieważ tylko prawidłowa klasa ma wartość 1, wzór ostatecznie również sprowadza się do:
\[
L=-\log(p_y)
\]

Binary Crossentropy

Dla klasyfikacji binarnej wzór wygląda następująco:

L=-y\(log(p)+(1-y)\log(1-p)
      
gdzie:
-\(y\) — prawidłowa etykieta: 0 albo 1,
-\(p\) — przewidywane prawdopodobieństwo klasy 1,
-\(L\) — wartość błędu.

    
Dla wielu obserwacji oblicza się najczęściej średnią wartość tych błędów.

Parameter from_logits przekazany w przykładzie do BinaryCrossentropy informuje algorytm, czy w ostatniej warstwie naszego modelu użyta została funkcja aktywacyjnej, czy nie. Podanie wartości True jest bardziej stabilne numerycznie i polecane. Jeśli ją ustawimy, nie możemy używać w ostatniej warstwie naszej sieci funkcji aktywacyjnej.
