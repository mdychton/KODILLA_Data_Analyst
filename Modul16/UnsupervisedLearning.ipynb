Widać tutaj, iż mamy kilka tzw. foldów, które są poszczególnymi podziałami naszych danych. Jeden fold można porównać do jednego podziału dokonanego przez train_test_split.

Przeanalizujmy wykres, aby zrozumieć ten mechanizm:

dane dzielimy na treningowe i testowe,

normalnie, gdybyśmy podzielili je metodą train_test_split na dane treningowe i walidacyjne, czyli

#->>>>>>>>>>>>>Różnica pomiędzy poszczególnymi foldami polega na tym, że w każdym z nich bierzemy jedną część jako zbiór testowy, pozostałe cztery jako zbiór treningowy.

Z tym że zbiór testowy zawsze jest różny, dla każdego folda jest inny.

Przeanalizujmy więc wszystkie foldy:

w pierwszym foldzie bierzemy Fold 1 (pierwszą część, 20% całego zbioru jako dane testowe, pozostałe 2, 3, 4, 5 części, jako dane treningowe)
w drugim foldzie bierzemy Fold 2 (druga część jako dane testowe (20%), pozostałe 1, 3, 4, 5 jako dane treningowe)
...
Widzimy tutaj, iż zbiór, na którym testujemy poszczególne foldy, zawsze jest inny, nigdy się nie powtarza.

Metoda ta pozwala nam na uniknięcie sytuacji, w której pomniejszymy swój zbiór treningowy o jakieś dane, gdyż trenujemy nasze modele (tak, wiele modeli, o czym zaraz się przekonamy) wszystkich danych (każdy na części całego zbioru, ale sumarycznie używamy wszystkich danych).

Ponadto sprawdzamy model na różnych danych testowych, czyli niewidzianych podczas szkolenia modelu. Mamy więc większą pewność, iż w naszym podziale nie ma wzorców, które powtarzają się zarówno w zbiorze treningowym i testowym, co mogłoby sztucznie podbić naszą metrykę.

```python
from sklearn.model_selection import KFold

kf = KFold(n_splits=5, shuffle=True, random_state=10)
```

# wazne

Metoda KFold i wszystkie z tej rodziny zwracają indeksy danych do testowania i trenowania, nie same podzielone dane. Należy więc użyć tych indeksów, aby wyciągnąć interesujące nas dane za pomocą [] albo operatora .iloc w przypadku Pandas Dataframe.

Trzeba to zrobić zarówno dla danych wejściowych, jak i wyjściowych.

```python
for fold_nr, (train_idx, test_idx) in enumerate(kf.split(X)):
  X_train = X[train_idx]
  X_test = X[test_idx]

  y_train = y[train_idx]
  y_test = y[test_idx]
```

Użycie enumerate pozwala uzyskać numer aktualnego foldu.

Ponieważ chcemy uzyskać niezależną ocenę naszego modelu dla każdego foldu, tworzymy osobny model i osobno go trenujemy i ewaluujemy. Wytrenowane modele możemy zapisać do tablicy czy słownika, aby mieć do nich dostęp w przyszłości.

Teraz wewnątrz foldu wytrenuj model na danych treningowych, określ jego wynik na danych testowych, zapisz wynik i model w odpowiednich tablicach.

```python
import numpy as np  # biblioteka do obliczeń numerycznych i pracy z tablicami

from sklearn.model_selection import KFold  # narzędzie do walidacji krzyżowej K-Fold
from sklearn.preprocessing import StandardScaler  # narzędzie do standaryzacji cech
from sklearn.linear_model import SGDClassifier  # klasyfikator uczony metodą SGD

scaler = StandardScaler()  # utworzenie obiektu skalującego

kf = KFold(n_splits=5, shuffle=True, random_state=1)  # podział danych na 5 foldów z losowym mieszaniem

models = []  # lista na wytrenowane modele
scores = []  # lista na wyniki poszczególnych foldów

for fold_nr, (train_idx, test_idx) in enumerate(kf.split(X)):  # iteracja po kolejnych foldach

    X_train = X[train_idx]  # wybór obserwacji treningowych
    X_test = X[test_idx]  # wybór obserwacji testowych

    y_train = y[train_idx]  # wybór etykiet treningowych
    y_test = y[test_idx]  # wybór etykiet testowych

    X_train = scaler.fit_transform(X_train)  # dopasowanie skalera i standaryzacja danych treningowych
    X_test = scaler.transform(X_test)  # standaryzacja danych testowych tym samym scalerem

    clf = SGDClassifier(random_state=1).fit(X_train, y_train)  # utworzenie i wytrenowanie modelu

    models.append(clf)  # zapisanie modelu do listy
    scores.append(clf.score(X_test, y_test))  # obliczenie accuracy i zapisanie wyniku

print("wyniki poszczególnych foldów: ", scores)  # wyświetlenie wyników dla każdego foldu
print("średni wynik wszystkich foldów: ", np.array(scores).mean())  # wyświetlenie średniego wyniku walidacji
```

# tlumaczenie 


Kod wykorzystuje 5-krotną walidację krzyżową (K-Fold Cross Validation) do oceny modelu SGDClassifier.

Dane są dzielone na 5 części. W każdej iteracji model jest trenowany na 4 częściach danych, a testowany na pozostałej części. Przed trenowaniem wykonywana jest standaryzacja cech za pomocą StandardScaler. Dla każdego foldu obliczana jest skuteczność modelu (accuracy), a na końcu wyświetlane są wyniki dla wszystkich foldów oraz ich średnia.

Dzięki temu można uzyskać bardziej wiarygodną ocenę jakości modelu niż przy pojedynczym podziale na zbiór treningowy i testowy.

# Wrapper

Po przejściu przez mozolny sposób użycia kroswalidacji, pora na wygodniejszy wrapper obecny w bibliotece sklearn. Działa on jednak tylko dla modeli z biblioteki sklearn, dla sieci neuronowych trzeba przejść przez cały proces opisany powyżej.

Użyjemy dwóch metod:

- cross_val_score (opis w dokumentacji) do wyciągania metryki klasyfikatora,

- cross_val_predict (opis w dokumentacji) do predykcji wyjść klasyfikatora.

Obydwie te metody działają w zupełnie w taki sam sposób, jak ta napisana przez nas z użyciem KFold, czyli

- robią kopie klasyfikatora wewnątrz siebie i wewnątrz każdego folda oraz zerują jego parametry do stanu początkowego,
- z tego powodu nie trzeba na nim wywoływać metody .fit() przed użyciem tych metod,
- metody te są używane głównie w celach sprawdzenia jak dany zestaw parametrów działa, gdyż nie zwracają nam klasyfikatorów. Z tego też powodu mogą być niereprodukowalne, jeśli nie ustawimy random_state,
- jeśli chcemy mieć dostęp do wszystkich klasyfikatorów, które zostały użyte podczas kroswalidacji, musimy użyć metody z KFold.
  
Użycie metod cross_val_predict i cross_val_score w celu sprawdzenia modelu:

```python
from sklearn.model_selection import cross_val_score, cross_val_predict  # import funkcji do walidacji krzyżowej

# stworzenie klasyfikatora
clf = SGDClassifier(random_state=1)  # utworzenie modelu SGD z ustalonym ziarnem losowości

# użycie metody cross_val_score do sprawdzenia
# działania naszego modelu na różnych podziałach
cv_score = cross_val_score(clf, X, y, cv=5)  # wykonanie 5-krotnej walidacji krzyżowej i zapis wyników

print("wynik kroswalidacji: ", cv_score)  # wyświetlenie wyników dla poszczególnych foldów
print("średni wynik wszystkich foldów: ", cv_score.mean())  # wyświetlenie średniego wyniku ze wszystkich foldów
```

Kod tworzy model SGDClassifier i ocenia jego skuteczność przy użyciu funkcji cross_val_score(). Funkcja automatycznie wykonuje 5-krotną walidację krzyżową, czyli pięciokrotnie dzieli dane na część treningową i testową, trenuje model oraz oblicza jego wynik. Na końcu wyświetlane są wyniki dla każdego foldu oraz ich średnia, która stanowi ogólną ocenę jakości modelu. Dzięki cross_val_score() cały proces walidacji krzyżowej jest realizowany automatycznie, bez konieczności pisania własnej pętli.

Mimo że kroswalidacja jest takim świetnym narzędziem, zarówno ona, jak i train_test_split, posiadają jedna wspólną niedoskonałość:

- podczas dzielenia danych istnieje możliwość, że w danych treningowych uzyskamy samą klasę A, natomiast w danych testowych samą klasę B
    
- nieważne ile foldów zrobimy, zawsze istnieje pewne ryzyko, że uzyskamy taki podział

Zobaczmy teraz, jak to działa na przykładzie. Poniżej znajduje się kod z podziałem za pomocą train_test_split bez użycia stratyfikacji:

```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=1000,
    n_classes=2,
    weights=[0.99, 0.01],
    flip_y=0,
    random_state=1
)

X_train, X_test, y_train, y_test = \
    train_test_split(X, y, test_size = 0.2, random_state=3)

train_0, train_1 = len(y_train[y_train==0]), len(y_train[y_train==1])
test_0, test_1 = len(y_test[y_test==0]), len(y_test[y_test==1])

print('>Train: 0=%d, 1=%d, Test: 0=%d, 1=%d' % (train_0, train_1, test_0, test_1))
```

Widać więc, że w danych treningowych są wszystkie klasy typu 1, natomiast nie ma ich w danych testowych. W fazie testowania naszego modelu nie będziemy mieli więc informacji, jak nasz model będzie się sprawował dla takiego typu danych – jest to Złe, przez duże Z.

Przepisanie powyższego przykładu, aby dane były rozłożone równomiernie w zbiorze treningowych i testowym:




```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=1000,
    n_classes=2,
    weights=[0.99, 0.01],
    flip_y=0,
    random_state=1
)

X_train, X_test, y_train, y_test = \
    train_test_split(X, y, test_size = 0.2, random_state=3, stratify=y)

train_0, train_1 = len(y_train[y_train==0]), len(y_train[y_train==1])
test_0, test_1 = len(y_test[y_test==0]), len(y_test[y_test==1])

print('>Train: 0=%d, 1=%d, Test: 0=%d, 1=%d' % (train_0, train_1, test_0, test_1))
```

Pojawia się tutaj parametr stratify, zapewniający prawidłowy rozkład danych.

Widzimy, że przy podziale 80%/20%, 80% obiektów klasy 1 znajduje się w zbiorze treningowym, natomiast 20% w zbiorze testowym i tak powinno być.

```python
import numpy as np  # biblioteka do obliczeń numerycznych i pracy z tablicami

from sklearn.model_selection import KFold  # narzędzie do walidacji krzyżowej K-Fold
from sklearn.preprocessing import StandardScaler  # narzędzie do standaryzacji cech
from sklearn.linear_model import SGDClassifier  # klasyfikator uczony metodą SGD

scaler = StandardScaler()  # utworzenie obiektu skalującego

kf = KFold(n_splits=5, shuffle=True, random_state=1)  # podział danych na 5 foldów z losowym mieszaniem

models = []  # lista na wytrenowane modele
scores = []  # lista na wyniki poszczególnych foldów

for fold_nr, (train_idx, test_idx) in enumerate(kf.split(X, y)):

    X_train = X[train_idx]  # wybór obserwacji treningowych
    X_test = X[test_idx]  # wybór obserwacji testowych

    y_train = y[train_idx]  # wybór etykiet treningowych
    y_test = y[test_idx]  # wybór etykiet testowych

    X_train = scaler.fit_transform(X_train)  # dopasowanie skalera i standaryzacja danych treningowych
    X_test = scaler.transform(X_test)  # standaryzacja danych testowych tym samym scalerem

    clf = SGDClassifier(random_state=1).fit(X_train, y_train)  # utworzenie i wytrenowanie modelu

    models.append(clf)  # zapisanie modelu do listy
    scores.append(clf.score(X_test, y_test))  # obliczenie accuracy i zapisanie wyniku

print("wyniki poszczególnych foldów: ", scores)  # wyświetlenie wyników dla każdego foldu
print("średni wynik wszystkich foldów: ", np.array(scores).mean())  # wyświetlenie średniego wyniku walidacji
```

```python
import numpy as np  # biblioteka do obliczeń numerycznych i pracy z tablicami

from sklearn.datasets import make_blobs  # funkcja do generowania sztucznych danych klastrowych

# środki naszych klastrów
centroids = np.array([  # zdefiniowanie współrzędnych środków klastrów
    [0.8, 2.0],  # środek klastra 1
    [-0.5, 2.0],  # środek klastra 2
    [-2.0, 2.0],  # środek klastra 3
    [-2.5, 2.5],  # środek klastra 4
    [-2.5, 1.0]   # środek klastra 5
])

# wprowadzenie szumu do naszych klastrów, aby rozrzucić próbki
blob_std = np.array([0.4, 0.3, 0.1, 0.1, 0.1])  # odchylenia standardowe dla poszczególnych klastrów

# stworzenie zbioru danych
X, y = make_blobs(
    n_samples=3000,  # liczba generowanych obserwacji
    centers=centroids,  # współrzędne środków klastrów
    cluster_std=blob_std,  # poziom rozproszenia punktów wokół środków
    random_state=7  # ziarno losowości zapewniające powtarzalność wyników
)
```

Kod generuje sztuczny zbiór danych do analizy skupień (clusteringu) przy użyciu funkcji make_blobs(). Zdefiniowanych zostaje 5 klastrów o określonych środkach (centroids) oraz różnym poziomie rozproszenia punktów (blob_std). Następnie tworzonych jest 3000 obserwacji rozmieszczonych wokół tych środków. Wynik zapisywany jest w zmiennej X (cechy punktów) oraz y (etykiety klastrów przypisane przez generator). Taki zbiór danych jest często wykorzystywany do testowania algorytmów grupowania, np. K-Means.

# I TU UZYCIE K-Means

```python
import numpy as np  # biblioteka do obliczeń numerycznych
import matplotlib.pyplot as plt  # biblioteka do tworzenia wykresów
from sklearn.cluster import KMeans  # import algorytmu K-Means

# funkcja do wizualizacji granic klastrów
def plot_decision_boundaries(model, X):

    x_min = X[:, 0].min() - 0.5  # minimalna wartość osi X
    x_max = X[:, 0].max() + 0.5  # maksymalna wartość osi X

    y_min = X[:, 1].min() - 0.5  # minimalna wartość osi Y
    y_max = X[:, 1].max() + 0.5  # maksymalna wartość osi Y

    step = (x_max - x_min) / 500  # dynamiczne zagęszczenie siatki

    xx, yy = np.meshgrid(
        np.arange(x_min, x_max, step),  # punkty siatki dla osi X
        np.arange(y_min, y_max, step)   # punkty siatki dla osi Y
    )

    grid = np.c_[xx.ravel(), yy.ravel()]  # połączenie współrzędnych siatki

    Z = model.predict(grid)  # przypisanie klastra do każdego punktu siatki
    Z = Z.reshape(xx.shape)  # dopasowanie wyników do kształtu siatki

    plt.contourf(xx, yy, Z, alpha=0.3)  # narysowanie obszarów klastrów

    plt.scatter(
        X[:, 0],
        X[:, 1],
        c=model.labels_,
        s=10
    )  # wyświetlenie punktów danych

    plt.scatter(
        model.cluster_centers_[:, 0],
        model.cluster_centers_[:, 1],
        marker='x',
        s=200
    )  # wyświetlenie centroidów klastrów


clf = KMeans(n_clusters=5, random_state=1)  # utworzenie modelu K-Means z 5 klastrami

y_pred = clf.fit_predict(X)  # dopasowanie modelu i przypisanie punktów do klastrów

plt.figure(figsize=(10, 5))  # utworzenie wykresu o rozmiarze 10x5 cali

plot_decision_boundaries(clf, X)  # wizualizacja granic pomiędzy klastrami

plt.title('K-Means Clustering')  # dodanie tytułu wykresu
plt.show()  # wyświetlenie wykresu
```

```python
import pandas as pd  # biblioteka do tworzenia tabel
from scipy.spatial.distance import cdist  # funkcja do liczenia odległości między punktami

found_centroids = clf.cluster_centers_  # centroidy wyznaczone przez model K-Means

distances = cdist(centroids, found_centroids)  # obliczenie odległości między centroidami prawdziwymi i wyznaczonymi

closest = distances.argmin(axis=1)  # znalezienie najbliższego centroidu modelu dla każdego prawdziwego centroidu

comparison = pd.DataFrame({  # utworzenie tabeli porównawczej
    'Wartości prawdziwe': [list(np.round(c, 2)) for c in centroids],  # zaokrąglenie prawdziwych centroidów
    'Wartości wyznaczone': [list(np.round(found_centroids[i], 8)) for i in closest]  # zaokrąglenie centroidów znalezionych przez model
})

print(comparison)  # wyświetlenie tabeli
```

Jako wynik działania naszego algorytmu otrzymujemy przypisanie poszczególnych próbek do poszczególnych grup/klastrów. Spróbujmy teraz zwizualizować, jak nasz algorytm dokonał przypisania. Do tej i przyszłych wizualizacji, będziemy używać metody plot_decision_boundaries znajdującej się w notatniku.

# Zasada działania KMeans

- 1 algorytm wybiera losowo centroidy, czyli środki poszczególnych grup,
- 2 przypisuje punkty do grup, biorąc pod uwagę najmniejszą odległość od centroidy,
- 3 poprawia pozycję centroid tak, aby najbardziej pasowały do danych,
- 4 powtarza od punktu 2 aż do momentu, kiedy centroidy przestaną się ruszać,
- 5 algorytm wykorzystuje wybór losowych centroidów, zatem algorytm jest uruchamiany kilkukrotnie, żeby zweryfikować najczęstszą zbieżność.

```python
import numpy as np  # biblioteka do obliczeń numerycznych
import matplotlib.pyplot as plt  # biblioteka do tworzenia wykresów
from sklearn.cluster import KMeans  # import algorytmu K-Means

# funkcja do wizualizacji granic klastrów
def plot_decision_boundaries(model, X):

    x_min = X[:, 0].min() - 0.5  # minimalna wartość osi X
    x_max = X[:, 0].max() + 0.5  # maksymalna wartość osi X

    y_min = X[:, 1].min() - 0.5  # minimalna wartość osi Y
    y_max = X[:, 1].max() + 0.5  # maksymalna wartość osi Y

    step = (x_max - x_min) / 500  # zagęszczenie siatki

    xx, yy = np.meshgrid(
        np.arange(x_min, x_max, step),  # punkty siatki dla osi X
        np.arange(y_min, y_max, step)   # punkty siatki dla osi Y
    )

    grid = np.c_[xx.ravel(), yy.ravel()]  # utworzenie siatki punktów

    Z = model.predict(grid)  # przypisanie klastra do każdego punktu siatki
    Z = Z.reshape(xx.shape)  # dopasowanie do kształtu siatki

    plt.contourf(xx, yy, Z, alpha=0.3)  # narysowanie granic klastrów

    plt.scatter(
        X[:, 0],
        X[:, 1],
        c=model.labels_,
        s=10
    )  # wyświetlenie punktów danych

    plt.scatter(
        model.cluster_centers_[:, 0],
        model.cluster_centers_[:, 1],
        marker='x',
        s=200
    )  # wyświetlenie centroidów


# wizualizacja pierwszych trzech iteracji algorytmu
for max_iter in [1, 2, 3, 300]:

    clf = KMeans(
        n_clusters=5,       # liczba klastrów
        init="random",      # losowa inicjalizacja centroidów
        n_init=1,           # tylko jedno uruchomienie algorytmu
        algorithm="lloyd",  # klasyczny algorytm Lloyda
        max_iter=max_iter,  # maksymalna liczba iteracji
        random_state=42     # ustalone ziarno losowości
    )

    clf.fit(X)  # trening modelu

    plt.figure(figsize=(10, 5))  # utworzenie wykresu

    plot_decision_boundaries(clf, X)  # wizualizacja podziału

    plt.title(f'K-Means po {max_iter} iteracji')  # tytuł wykresu

    plt.show()  # wyświetlenie wykresu
```

```python
# tworzymy modele dla różnych wartości k
kmeans_per_k = [                                            # lista modeli K-Means dla różnych wartości k
    KMeans(                                                 # utworzenie modelu K-Means
        n_clusters=k,                                       # liczba klastrów
        random_state=42                                     # ustalenie ziarna losowości dla powtarzalności wyników
    ).fit(X)                                                # trenowanie modelu na danych X
    for k in range(2, 20)                                   # wartości k od 2 do 19
]
```

Pierwszą metodą jest graficzny wybór liczby klastrów w oparciu o metodę łokcia – elbow method. Jest to wykres SSE (Sum of Squared Errors) w zależności od liczby klastrów. SSE to metryka minimalizowana wewnątrz klastrów, jest to suma różnic pomiędzy środkiem klastra a obserwacją należącą do danego klastra. Wraz z kolejnymi klastrami odległości pomiędzy obserwacjami a klastrami będą mniejsze, zatem też SSE będzie maleć. Na podstawie wykresu należy dobrać taki k, po którym nie widać znaczących zmian z kolejnymi klastrami.

```python
import matplotlib.pyplot as plt                             # biblioteka do tworzenia wykresów

inertias = [model.inertia_ for model in kmeans_per_k]       # pobranie wartości SSE (inertia) dla każdego modelu

plt.plot(range(2, 20), inertias, 'bx-')                     # wykres SSE w zależności od liczby klastrów

plt.xlabel('K')                                             # etykieta osi X - liczba klastrów

plt.ylabel('SSE')                                           # etykieta osi Y - suma kwadratów odległości od centroidów

plt.title('Elbow Method using SSE')                         # tytuł wykresu metody łokcia

plt.show()                                                  # wyświetlenie wykresu
```

```python
from sklearn.linear_model import LogisticRegression  # regresja logistyczna
from sklearn.datasets import load_digits  # zbiór danych cyfr
from sklearn.model_selection import train_test_split  # podział na zbiór treningowy i testowy
from sklearn.pipeline import Pipeline  # tworzenie potoku przetwarzania
from sklearn.cluster import KMeans  # algorytm grupowania K-Means

# wczytanie zbioru danych cyfr
X_digits, y_digits = load_digits(return_X_y=True)

# podział danych na zbiór treningowy i testowy
X_train, X_test, y_train, y_test = \
    train_test_split(
        X_digits,
        y_digits,
        random_state=2
    )

# utworzenie modelu regresji logistycznej
log_reg = LogisticRegression(
    max_iter=5000,       # maksymalna liczba iteracji
    random_state=2      # ustalenie ziarna losowości
)

# trenowanie modelu
log_reg.fit(X_train, y_train)

# obliczenie dokładności na zbiorze testowym
print("Bez KMeans :", log_reg.score(X_test, y_test))

# utworzenie potoku K-Means + Regresja Logistyczna
pipeline = Pipeline([
    (
        "kmeans",
        KMeans(
            n_clusters=40,  # liczba klastrów
            random_state=2  # ustalenie ziarna losowości
        )
    ),
    (
        "log_reg",
        LogisticRegression(
            max_iter=5000,      # maksymalna liczba iteracji
            random_state=2      # ustalenie ziarna losowości
        )
    ),
])

# trenowanie całego potoku
pipeline.fit(X_train, y_train)

# obliczenie dokładności modelu z K-Means
print("Z KMeans :", pipeline.score(X_test, y_test))
```

# opis

W tym przykładzie porównujemy skuteczność klasycznej regresji logistycznej z modelem wykorzystującym potok (Pipeline) zawierający najpierw algorytm K-Means, a następnie regresję logistyczną.

K-Means grupuje podobne obserwacje w klastry i przekształca dane do nowej reprezentacji. Następnie regresja logistyczna uczy się na tych przekształconych danych. Dzięki temu można sprawdzić, czy wcześniejsze grupowanie obserwacji poprawi jakość klasyfikacji.

Porównując wyniki:

Bez KMeans – regresja logistyczna działa bezpośrednio na oryginalnych cechach.
Z KMeans – regresja logistyczna korzysta z informacji o przynależności do klastrów wyznaczonych przez K-Means.

Jeżeli dokładność modelu z K-Means jest wyższa, oznacza to, że grupowanie danych pomogło wydobyć dodatkową strukturę ze zbioru danych i ułatwiło klasyfikację.

# DBSCAN (inny algorytm klasteryzacji)

Interesujący jest fakt, że o ile w przypadku dotychczas omawianych algorytmów, wszystkie miały metodę predict() do przewidywania wartości/klasy/klastra, do której należy nowa dana lub cała próbka, bazując na tym, czego się nauczyły – DBSCAN nie pozwala na to.



#  wazne

Może on być zastosowany tylko na kompletnych danych, które już mamy w całości. Jeśli dodamy nowe elementy, musimy jeszcze raz uczyć cały model.

Zasada działania tego algorytmu:

- dla każdego elementu zbioru algorytm liczy elementy w małym dystansie epsilon. Nazywa się to epsilon neighborhood.

- Jeżeli jest co najmniej min_samples elementów w sąsiedztwie danego punktu, w odległości epsilon od niego, mamy do czynienia z centrum naszego klastra.

- Wszystkie elementy wokoło tego centrum należą do danego klastra.

- WAŻNE. Wszystkie obiekty, które nie należą do żadnego klastra i nie mają w swoim otoczeniu (odległość epsilon) żadnego klastra, są anomaliami.

# ANOMALIE - Wszystkie obiekty, które nie należą do żadnego klastra i nie mają w swoim otoczeniu (odległość epsilon) żadnego klastra, są anomaliami.

# ZALETY
Algorytm ten dobrze działa, jeśli dane są gęsto rozmieszczone i jest wyraźna separacja między poszczególnymi klastrami.

Ponadto pozwala on odsiać wszystkie anomalie, czyli dane odbiegające wartościami od pozostałych.

Jest to przydatne dlatego, że podczas nauki modelu takie anomalie potrafią wiele namieszać i można spędzić długie godziny, poprawiając model.

```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons

# tworzymy zbiór moon
X, y = make_moons(n_samples=1000, noise=0.08)

# tworzymy pierwszy klasyfikator DBSCAN z eps = 0.05
dbscan = DBSCAN(eps=0.05, min_samples=5)
dbscan.fit(X)

# drugi DBSCAN z eps = 0.2 (większy epsilon, większa przestrzeń wokół)
dbscan_2 = DBSCAN(eps=0.2, min_samples=5)
dbscan_2.fit(X)
```

```python
import matplotlib.pyplot as plt  # biblioteka do tworzenia wykresów

plt.figure(figsize=(12, 5))  # utworzenie figury o szerokości 12 i wysokości 5

# DBSCAN z eps=0.05
plt.subplot(1, 2, 1)  # pierwszy wykres w układzie 1 wiersz, 2 kolumny

plt.scatter(
    X[:, 0],           # wartości pierwszej cechy na osi X
    X[:, 1],           # wartości drugiej cechy na osi Y
    c=dbscan.labels_,  # kolor punktu zależny od przypisanego klastra
    s=10               # rozmiar punktów
)

plt.title('DBSCAN eps=0.05')  # tytuł pierwszego wykresu

# DBSCAN z eps=0.2
plt.subplot(1, 2, 2)  # drugi wykres w układzie 1 wiersz, 2 kolumny

plt.scatter(
    X[:, 0],             # wartości pierwszej cechy na osi X
    X[:, 1],             # wartości drugiej cechy na osi Y
    c=dbscan_2.labels_,  # kolor punktu zależny od przypisanego klastra
    s=10                 # rozmiar punktów
)

plt.title('DBSCAN eps=0.2')  # tytuł drugiego wykresu

plt.show()  # wyświetlenie wykresów
```

# opis
Przy eps=0.05 zobaczysz zwykle więcej małych klastrów i więcej punktów oznaczonych jako szum. Przy eps=0.2 dwa "księżyce" powinny zostać połączone w dwa duże klastry z niewielką liczbą punktów szumu.

# Gaussian Mixture Models (GNN)

Mieliśmy już modele, które bazowały na:

- gęstości punktów (DBSCAN),
- przesuwaniu centroid i wyznaczaniu odległości od punktów (KMeans).
  
Teraz pora na trzecią rodzinę algorytmów do klasteryzacji, bazującą na rozkładzie prawdopodobieństwa, a dokładniej na rozkładzie Gaussa (czym jest sam rozkład Gaussa, nie jest w tym momencie istotne, ale w razie czego wujek Google pomoże ;))

Zobaczmy, jak taki podział wygląda w praktyce. Będziemy bazowali na elipsoidalnych zbiorach danych (w kształcie "bagietki") stworzonych za pomocą poniższego kodu:

```python
from sklearn.datasets import make_blobs  # generowanie sztucznych danych
from sklearn.mixture import GaussianMixture  # model Gaussian Mixture Model
import numpy as np  # biblioteka do obliczeń numerycznych
import matplotlib.pyplot as plt  # biblioteka do tworzenia wykresów

# ==================================================
# Generowanie danych
# ==================================================

# tworzymy dwa klastry danych
X1, y1 = make_blobs(
    n_samples=800,                    # liczba obserwacji
    centers=((2, -2), (-2, 2)),       # położenie centrów klastrów
    random_state=2                    # ustalenie ziarna losowości
)

# transformacja danych powodująca powstanie wydłużonych klastrów (elips)
X1 = X1.dot(
    np.array([
        [0.374, 0.95],
        [0.732, 0.598]
    ])
)

# tworzymy trzeci klaster
X2, y2 = make_blobs(
    n_samples=200,    # liczba obserwacji
    centers=1,        # jeden klaster
    random_state=42   # ustalenie ziarna losowości
)

# przesuwamy trzeci klaster w inne miejsce przestrzeni
X2 = X2 + [6, -8]

# łączymy oba zbiory danych w jedną macierz cech
X = np.r_[X1, X2]

# łączymy etykiety w jeden wektor
y = np.r_[y1, y2]

# ==================================================
# Funkcja do wizualizacji klastrów GMM
# ==================================================

def plot_gaussian_mixture(clf, X, resolution=300):

    # wyznaczenie zakresu osi X
    x_min = X[:, 0].min() - 1

    # wyznaczenie zakresu osi X
    x_max = X[:, 0].max() + 1

    # wyznaczenie zakresu osi Y
    y_min = X[:, 1].min() - 1

    # wyznaczenie zakresu osi Y
    y_max = X[:, 1].max() + 1

    # utworzenie siatki punktów
    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, resolution),
        np.linspace(y_min, y_max, resolution)
    )

    # zamiana siatki na listę współrzędnych
    grid = np.c_[xx.ravel(), yy.ravel()]

    # przewidywanie klastra dla każdego punktu siatki
    Z = clf.predict(grid)

    # dopasowanie wyniku do kształtu siatki
    Z = Z.reshape(xx.shape)

    # rysowanie granic klastrów
    plt.contourf(
        xx,
        yy,
        Z,
        alpha=0.3
    )

    # przewidywanie klastra dla każdej obserwacji
    labels = clf.predict(X)

    # rysowanie punktów danych
    plt.scatter(
        X[:, 0],       # współrzędne X
        X[:, 1],       # współrzędne Y
        c=labels,      # kolor zależny od klastra
        s=10           # rozmiar punktów
    )

# ==================================================
# Trenowanie modelu GMM
# ==================================================

gm = GaussianMixture(
    n_components=3,   # liczba poszukiwanych klastrów
    n_init=100,        # liczba uruchomień algorytmu
    random_state=42   # ustalenie ziarna losowości
)

# trening modelu na danych
gm.fit(X)

# ==================================================
# Wizualizacja wyników
# ==================================================

plt.figure(figsize=(12, 6))  # utworzenie figury

# narysowanie klastrów i granic decyzyjnych
plot_gaussian_mixture(gm, X)

# dodanie tytułu wykresu
plt.title("Gaussian Mixture Model")

# wyświetlenie wykresu
plt.show()
```

Elipsoidalne kształty w różnych kolorach są właśnie wartościami rozkładu Gaussa, czyli z jakim prawdopodobieństwem dany element należy do danego klastra. Im ciemniejszy kolor, tym bardziej jesteśmy pewni.

# WAZNE

Podobnie, jak w przypadku KMeans, algorytm ten może zbiegać do złych rozwiązań, dlatego tu także istnieje parametr n_init, który określa nam ilość prób, jakie algorytm ma wykonać.

W KMeans domyślna wartość n_init to 10, więc nie musimy jej zmieniać. W GMM domyślna wartość to 1, więc trzeba ją zmienić na większą, inaczej algorytm w 99% przypadków da złe wyniki.

Można łatwo sprawdzić, czy algorytm skończył działanie za pomocą poniższej metody:

gm.coverged_
Podobnie można sprawdzić, ile iteracji potrzebował:

gm.n_iter_
Podobnie, jak w przypadku KMeans, w GMM również trzeba określić ilość klastrów, jaką chcemy otrzymać. Służy do tego parametr n_components.

Jeżeli chcemy sprawdzić, do jakiej grupy należą nowe elementy, możemy użyć standardowej metody predict().

gm.predict(X_test)
W przypadku większej ilości wymiarów (cech), bardziej skomplikowanych danych, np. tekstowych, algorytm może mieć problem z osiągnięciem optymalnego rozwiązania i będzie krążyć w nieskończoność.

Można rozwiązać ten problem za pomocą hiperparametru covariance_type. Przyjmuje on jedną z czterech wartości:

- full – bazowa wartość, stara się dopasować wszystkie klastry osobno, bez żadnego wzoru (np. klaster wygląda jak bagietka),
- tied – wszystkie klastry będą miały taki sam kształt, "bagietka" skierowana w taką samą stronę,
- diag – klastry mogą mieć kształty tylko elipsoidalne,
- spherical – wszystkie klastry to sfery (okręgi, koła).

```python
from sklearn.datasets import make_blobs  # generowanie sztucznych danych
from sklearn.mixture import GaussianMixture  # model Gaussian Mixture Model
import numpy as np  # biblioteka do obliczeń numerycznych
import matplotlib.pyplot as plt  # biblioteka do tworzenia wykresów

# ==================================================
# Generowanie danych
# ==================================================

# tworzymy dwa klastry danych
X1, y1 = make_blobs(
    n_samples=800,                    # liczba obserwacji
    centers=((2, -2), (-2, 2)),       # położenie centrów klastrów
    random_state=2                    # ustalenie ziarna losowości
)

# transformacja danych powodująca powstanie wydłużonych klastrów (elips)
X1 = X1.dot(
    np.array([
        [0.374, 0.95],
        [0.732, 0.598]
    ])
)

# tworzymy trzeci klaster
X2, y2 = make_blobs(
    n_samples=200,    # liczba obserwacji
    centers=1,        # jeden klaster
    random_state=42   # ustalenie ziarna losowości
)

# przesuwamy trzeci klaster w inne miejsce przestrzeni
X2 = X2 + [6, -8]

# łączymy oba zbiory danych w jedną macierz cech
X = np.r_[X1, X2]

# łączymy etykiety w jeden wektor
y = np.r_[y1, y2]

# ==================================================
# Funkcja do wizualizacji klastrów GMM
# ==================================================

def plot_gaussian_mixture(clf, X, resolution=300):

    # wyznaczenie zakresu osi X
    x_min = X[:, 0].min() - 1

    # wyznaczenie zakresu osi X
    x_max = X[:, 0].max() + 1

    # wyznaczenie zakresu osi Y
    y_min = X[:, 1].min() - 1

    # wyznaczenie zakresu osi Y
    y_max = X[:, 1].max() + 1

    # utworzenie siatki punktów
    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, resolution),
        np.linspace(y_min, y_max, resolution)
    )

    # zamiana siatki na listę współrzędnych
    grid = np.c_[xx.ravel(), yy.ravel()]

    # przewidywanie klastra dla każdego punktu siatki
    Z = clf.predict(grid)

    # dopasowanie wyniku do kształtu siatki
    Z = Z.reshape(xx.shape)

    # rysowanie granic klastrów
    plt.contourf(
        xx,
        yy,
        Z,
        alpha=0.3
    )

    # przewidywanie klastra dla każdej obserwacji
    labels = clf.predict(X)

    # rysowanie punktów danych
    plt.scatter(
        X[:, 0],       # współrzędne X
        X[:, 1],       # współrzędne Y
        c=labels,      # kolor zależny od klastra
        s=10           # rozmiar punktów
    )

# ==================================================
# Trenowanie modelu GMM
# ==================================================

gm = GaussianMixture(
    n_components=3,   # liczba poszukiwanych klastrów
    covariance_type= 'spherical',
    n_init=50,        # liczba uruchomień algorytmu
    random_state=42   # ustalenie ziarna losowości
)

# trening modelu na danych
gm.fit(X)

# ==================================================
# Wizualizacja wyników
# ==================================================

plt.figure(figsize=(12, 6))  # utworzenie figury

# narysowanie klastrów i granic decyzyjnych
plot_gaussian_mixture(gm, X)

# dodanie tytułu wykresu
plt.title("Gaussian Mixture Model")

# wyświetlenie wykresu
plt.show()
```

# GMM jako wykrywacz anomalii

GMM jest często używany jako detektor anomalii. To proste, ale działa pod jednym warunkiem – jeśli wiemy, jaki odsetek wartości chcemy filtrować.

Dla przypomnienia: anomalie to wartości, które odbiegają od naszego ogólnego obrazu danych, czyli np. znajdują się daleko od innych punktów.

Dlaczego odsiewamy anomalie?

Dlatego, że psują nam model. Przypomnijmy sobie, iż nasz model chce dopasować się do wszystkich punktów. Przyjmijmy, że mamy takie:



(1,2)
(1,3)
(1,0)
(1234324,34324234324)

#jak pozbyc sie anomalii

Ostatni punkt znacznie odbiega od pozostałych, czyli jest anomalią, która zaburza nam ogólny obraz naszego zbioru. Można to rozwiązać na dwa sposoby:

- skalowanie danych,
- odsianie anomalii, czyli usunięcie ich ze zbioru.

```python
from sklearn.datasets import make_blobs  # generowanie sztucznych danych
from sklearn.mixture import GaussianMixture  # model Gaussian Mixture Model
import numpy as np  # biblioteka do obliczeń numerycznych
import matplotlib.pyplot as plt  # biblioteka do tworzenia wykresów

# ==================================================
# Generowanie danych
# ==================================================

# tworzymy dwa klastry danych
X1, y1 = make_blobs(
    n_samples=800,                    # liczba obserwacji
    centers=((2, -2), (-2, 2)),       # położenie centrów klastrów
    random_state=2                    # ustalenie ziarna losowości
)

# transformacja danych powodująca powstanie wydłużonych klastrów (elips)
X1 = X1.dot(
    np.array([
        [0.374, 0.95],
        [0.732, 0.598]
    ])
)

# tworzymy trzeci klaster
X2, y2 = make_blobs(
    n_samples=200,    # liczba obserwacji
    centers=1,        # jeden klaster
    random_state=42   # ustalenie ziarna losowości
)

# przesuwamy trzeci klaster w inne miejsce przestrzeni
X2 = X2 + [6, -8]

# łączymy oba zbiory danych w jedną macierz cech
X = np.r_[X1, X2]

# łączymy etykiety w jeden wektor
y = np.r_[y1, y2]

# ==================================================
# Funkcja do wizualizacji klastrów GMM
# ==================================================

def plot_gaussian_mixture(clf, X, resolution=300):

    # wyznaczenie zakresu osi X
    x_min = X[:, 0].min() - 1

    # wyznaczenie zakresu osi X
    x_max = X[:, 0].max() + 1

    # wyznaczenie zakresu osi Y
    y_min = X[:, 1].min() - 1

    # wyznaczenie zakresu osi Y
    y_max = X[:, 1].max() + 1

    # utworzenie siatki punktów
    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, resolution),
        np.linspace(y_min, y_max, resolution)
    )

    # zamiana siatki na listę współrzędnych
    grid = np.c_[xx.ravel(), yy.ravel()]

    # przewidywanie klastra dla każdego punktu siatki
    Z = clf.predict(grid)

    # dopasowanie wyniku do kształtu siatki
    Z = Z.reshape(xx.shape)

    # rysowanie granic klastrów
    plt.contourf(
        xx,
        yy,
        Z,
        alpha=0.3
    )

    # przewidywanie klastra dla każdej obserwacji
    labels = clf.predict(X)

    # rysowanie punktów danych
    plt.scatter(
        X[:, 0],       # współrzędne X
        X[:, 1],       # współrzędne Y
        c=labels,      # kolor zależny od klastra
        s=10           # rozmiar punktów
    )

# ==================================================
# Trenowanie modelu GMM
# ==================================================


gm = GaussianMixture(
    n_components=3,        # liczba poszukiwanych klastrów
    covariance_type='full',# sposób modelowania kształtu klastrów
    n_init=10,             # liczba uruchomień algorytmu
    random_state=42        # ustalenie ziarna losowości
)

# trening modelu na danych
gm.fit(X)

# ==================================================
# Wykrywanie anomalii
# ==================================================

# obliczenie gęstości (prawdopodobieństwa) dla każdej obserwacji
density = gm.score_samples(X)

# wyznaczenie progu odcinającego 5% najmniej prawdopodobnych obserwacji
threshold = np.percentile(
    density,
    5
)

# wybór obserwacji uznanych za anomalie
anomalie = X[
    density < threshold
]

# wyświetlenie liczby znalezionych anomalii
print("Liczba anomalii:", len(anomalie))
# ==================================================
# Wizualizacja wyników
# ==================================================

plt.figure(figsize=(12, 6))  # utworzenie figury

# narysowanie klastrów i granic decyzyjnych
plot_gaussian_mixture(gm, X)

# dodanie tytułu wykresu
plt.title("Gaussian Mixture Model")

# wyświetlenie wykresu
plt.show()
```
