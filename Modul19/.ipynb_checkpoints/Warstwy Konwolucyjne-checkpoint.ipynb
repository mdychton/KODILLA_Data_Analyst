Działanie CNN (Convolutional Neural Network – Konwolucyjne Sieci Neuronowe) spróbujemy przybliżyć na podstawie rozróżniania kluczy. Tak, po to są sieci konwolucyjne – aby klasyfikować obiekty na obrazach.

Ludzie są w stanie sklasyfikować i zlokalizować obiekt na zdjęciu, które są "zaszumione" innymi obiektami. Rzadko kiedy na jednym zdjęciu mamy jeden obiekt. Przypomnij sobie Twoje ostatnie zdjęcie z wakacji, oprócz Ciebie mogły znajdować się na nim również palmy, jacht czy znana budowla. Komputery nadal mają problemy z rozróżnianiem obiektów na zdjęciach ze względu na kąt widzenia lub zmienność samej klasy, która może się pojawić. Jak to rozumieć? Klasyczne przetwarzanie obrazu i wykrywanie obiektów odbywa się za pomocą działań na macierzy obrazu. Filtr rozmycia jest wykorzystywany do redukcji szumu w obrazie, natomiast wyodrębnienie obiektu z tła zapewnia progowanie. Proces ten przebiega zgodnie z wcześniej przygotowanym algorytmem. Takie podejście działać może na linii produkcyjnej, gdzie zawsze mamy jeden obiekt o tym samym kolorze, jak i również podłoże jest to samo.

Sieci konwolucyjne działają w sposób odwrotny do klasycznego przetwarzania obrazu opartego na algorytmie zdefiniowanym przez człowieka. Na podstawie danych treningowych CNN automatycznie wyodrębnia cechy, które później zostaną wykorzystane do klasyfikacji obiektów. Głębokie sieci neuronowe naśladują architekturę kory wzrokowej ssaków, wyodrębniając lokalne pola recepcyjne i łącząc się z filtrami. Lokalne pola recepcyjne stanowią oddzielną przestrzeń, w której poszukiwane są wzorce. Odpowiedzi z filtrów są wielokrotnie podpróbkowywane i ponownie filtrowane. Neurony w korze wzrokowej ssaków są zorganizowane warstwowo w celu przetwarzania obrazów. Niektóre z nich mają funkcję rozpoznawania cech lokalnych, takich jak krawędzie, przecinające się linie lub kształty.

Klucze używane w analizie mają określone cechy. Konwolucyjna sieć neuronowa może być użyta do określenia, czy klucz jest kwadratowy czy okrągły, jaki jest rodzaj uchwytu, dziurki lub jaki kształt ma profil i kolor zęba. Znowu to zrobiliśmy – sposób działania natury próbujemy przełożyć na działanie komputera.

Zobacz jaki rozmiar mają obrazy robione przez Twój aparat. Załóżmy, że to 1600 x 1600 (zapewne większą rozdzielczość mają zdjęcia z Twojego aparatu), co daje w sumie 2 560 000 pikseli! Ponadto obraz ma trzy kanały kolorów. Kolejnym problemem są obiekty znajdujące się w różnych częściach obrazu. Jeśli w zbiorze treningowym klucz znajdował się tylko w lewym lub prawym górnym rogu obrazu, sieć przyjmie, że może on być zlokalizowany tylko w tych obszarach.

W dodatku takie obrazy powinniśmy przekształcić do jednowymiarowego wektora, przez co następuje utrata informacji. Konwolucyjne sieci neuronowe są podobne do zwykłych sieci neuronowych. Neurony również mają wagi, które zmieniają się podczas uczenia, i mają funkcje aktywacji – te które już poznaliśmy. Wszystkie modele sieci neuronowych mają podobną architekturę, co możesz zobaczyć niżej. Architektura zakłada, że sieć otrzyma na wejściu obraz. Natomiast zwykłe sieci otrzymują na wejściu tylko dane w postaci jednowymiarowego wektora. Dzięki takiej architekturze konwolucyjne sieci neuronowe szybko się uczą, jak i również mogą być odporne na różnej wielkości obiekty, czy nawet przesunięcia i obroty.

# Warstwa konwolucyjna

Piksele z obrazu wejściowego nie są łączone z każdym neuronem ukrytym. Warstwy splotowe zawężają obszar obrazu wejściowego (obszar sprawdzania) dla każdego neuronu. Wielkość tego obszaru zależy od wielkości filtra, zwanego jądrem. Neurony w kolejnej warstwie łączą się tylko z innymi neuronami w poprzedniej warstwie. Jądro jest macierzą o wymiarach h x w x d i znajduje się w warstwie splotowej, służącej do wykrywania cennych cech obiektów. Litera d oznacza liczbę zastosowanych różnych filtrów, zatem w jednej warstwie splotowej jest wiele filtrów. Rozmiar filtra jest określany przez specjalistę. Zwykle jest to 3x3 lub 5x5 oraz liczba użytych filtrów. Jądro przesuwa się od lewego górnego do prawego dolnego rogu obrazu, a dla każdego punktu w przestrzeni elementy tablicy są mnożone i sumowane.

Sprawdźmy, jak mogą działać różne kernele (filtr) dla zdjęcia naszego klucza.

Kernel ma pewne wartości, te wartości to nic innego jak wagi sieci neuronowej. Konwolucyjne sieci neuronowe wykorzystują wiele filtrów w celu znalezienia cech obrazu, które pozwolą na klasyfikację obiektów. To, co najważniejsze, filtry dzięki sieci neuronowej powstają w sposób automatyczny – wcześniej takie filtry tworzyć musiał programista.

W architekturze CNN istnieje wiele jąder odpowiedzialnych za wydobywanie tych cech. Jądro krawędziowe służy do wyróżniania dużych różnic w wartościach pikseli. Jeśli piksel znajdujący się obok ma podobną intensywność, to po transformacji będzie czarny. Duża różnica między nimi spowoduje, że będzie on biały. Filtr jądra rozmycia rozmywa różnice w sąsiadujących pikselach, zmniejszając szum na obrazie. Dolne jądro Sobela wykrywa krawędzie w określonym kierunku, wskazując różnice w sąsiadujących pikselach. Filtr Emboss działa na podobnej zasadzie jak filtr Sobela, uwypuklając różnice pikseli w określonym kierunku, np. pod skosem. Ostatnim z prezentowanych filtrów jest wyostrzanie.

Powyższe jądra spełniają wymagane minimum, aby wyjaśnić mechanizm ich działania w splotowych sieciach neuronowych. Pojedyncze jądro wykrywa określoną cechę z konkretnego miejsca obrazu wejściowego. Wiele neuronów korzystających z tego samego jądra tworzy mapę cech. Mapa cech zmienia się podczas uczenia modelu za pomocą minimalizowanej funkcji straty. Warstwa splotu składa się z kilku map cech, a operacja splotu jest wykonywana niezależnie dla każdego filtra. Powstałe w ten sposób mapy cech są rozłączne. Wynik operacji splotu jest przekazywany do funkcji aktywacji ReLU. Na wyjściu warstwy splotu mapy cech łączą się i stają się ostatecznym wynikiem jej działania. Większa liczba jąder wpływa na liczbę wykrywanych cech, co wydłuża czas treningu modelu. Podczas operacji splotu usuwane są boczne piksele i następuje utrata informacji. Można tego uniknąć, jeśli obraz wejściowy zostanie otoczony pikselami o wartości 0 (padding) lub skopiowany zostanie najbliższy piksel. Pierwsze warstwy splotu specjalizują się w rozpoznawaniu małych cech. Kolejne są odpowiedzialne za wychwytywanie bardziej złożonych cech.

# Warstwa łącząca

Warstwa łącząca agreguje wynik wartości splotowej przez pobranie wartości maksymalnej z konkretnego obszaru (Max pooling). Warstwa ta ma także możliwość uśredniania (Average Pooling), minimalizowania lub przekształcania na inną funkcję. Warstwa łączenia (Pooling Layer) w tym przypadku gromadzi tylko dane przy użyciu określonej funkcji agregacji. Jej zadaniem jest zmniejszenie o połowę rozmiaru wysokości i szerokości macierzy, pozostawiając głębokość (kanały) nienaruszoną. Dzięki takiemu zabiegowi zmniejszamy ryzyko przetrenowania poprzez określenie najbardziej wpływowych cech warstwy poolingowej. Taka procedura pozwala na szybsze trenowanie modelu.

Podobnie jak w warstwie konwolucyjnej, neurony warstwy poolingowej łączą się z innymi neuronami poprzedniej warstwy, które znajdują się w obszarze ‘zainteresowania’. W powyższym przykładzie dodano warstwę MaxPooling o wymiarach 2x2 i szerokości 2. Jest ona przesuwana nad macierzą wejściową i dla każdego punktu w przestrzeni umieszczana jest maksymalna wartość w macierzy wyjściowej. Wysokość i szerokość danych wyjściowych będą dwukrotnie mniejsze, co czterokrotnie zmniejszy obszar wyjściowy. CNN daje najlepsze wyniki, gdy stosuje się naprzemiennie warstwy konwolucyjne i łączenia.

# Warstwa w pełni połączona

Wejściem do tej warstwy jest ostatnia warstwa łącząca. Zadaniem tej warstwy jest spłaszczenie macierzy wag w wektor wag. Kolejno jest on łączony z następnymi warstwami, jednak jest to już standardowa architektura sztucznej sieci neuronowej. Ostatnia warstwa jest odpowiedzialna za określenie prawdopodobieństwa przynależności do danej klasy. Jeśli mamy do czynienia z klasyfikacją wieloklasową, wykorzystywana jest funkcja aktywacji softmax. W przypadku klasyfikacji binarnej wykorzystywana jest funkcja sigmoidalna.

# Regularyzacja 

regularyzacja jest potrzebna aby model się nie przeuczył. Znasz już regularyzację jaką jest Lasso i Ridge, możesz je również zastosować w sieciach neuronowych. Sieci neuronowe posiadają zdecydowanie (MNÓSTWO – o czym się przekonasz niedługo) więcej parametrów niż typowe modele uczenia maszynowego. Daje to takim modelom dużą elastyczność w procesie uczenia. Niesie to ze sobą wiele pozytywnych efektów, takich jak dostosowywanie się do dowolnej funkcji. Wadą jest podatność sieci na przepełnienie. Aby ograniczyć to zjawisko i zwiększyć możliwość generalizacji, stosuje się techniki regularyzacji sieci.

# Early stopping – wczesne zatrzymanie

Sprawdzany jest błąd na zbiorze walidacyjnym. Ustawiany jest również hiperparametr cierpliwości patience na n epok. Jeśli po tej liczbie epok błąd się nie zmniejsza, to model sieci neuronowej przestaje być uczony. Metoda ta działa bardzo dobrze w praktyce, nie musisz wybierać czy lepszym wyborem będzie 5, 10 czy 1000 epok. Pamiętaj, zbiór walidacyjny to nie to samo co zbiór testowy.

# Dropout - odrzucenie

Ta metoda polega na losowym wyłączeniu neuronu. Podczas każdej epoki uczenia nie wykorzystywane są wszystkie neurony w procesie predykcji i propagacji wstecznej.

Porzucana jest procentowa liczba neuronów zgodnie z parametrem p. Zazwyczaj jest to 50%. Jednak nie jest to oczywiście złota reguła, w większości sytuacji działa dobrze, ale nie najlepiej – pamiętaj o tym. Dzięki porzuceniu każdy neuron staje się niezależny i użyteczny. Sieć staje się bardziej uniwersalna i ogólna. Po etapie uczenia każda waga neuronu jest mnożona przez 0,5, ponieważ podczas testowania każdy neuron jest połączony ze średnio dwa razy większą liczbą neuronów.

# Data augmentation – augmentacja danych

Jeśli mamy więcej danych, to również nasze modele są lepszej jakości. W niektórych jednak przypadkach jest to bardzo pracochłonne lub niemożliwe, aby uzyskać nowe dane. By ominąć ten problem stosowana jest technika augmentacji. Polega ona na zwiększeniu zbioru treningowego przez modyfikację już istniejących danych. Stosuje się różne techniki przekształceń, takie jak obrót, powiększanie, przesuwanie czy zmiana jasności obrazu. Model staje się bardziej wydajny, ponieważ w każdej epoce uczy się na innym zbiorze danych. Zbiór treningowy jest bogatszy i bardziej zróżnicowany w porównaniu z poprzednim. Wzbogacanie odbywa się w czasie rzeczywistym, a przekształcone dane nie są zapisywane. Zobacz, jak możemy przekształcić jedno zdjęcie klucza na kilka różnych.

# Transfer learning – wykorzystanie już nauczonego modelu

Wykorzystujemy już nauczoną sieć do naszego problemu. W związku z tym sieć nie uczy się od losowych wag, tylko zaczyna od wzorców nauczonych dla innego problemu. Przyjmuje się, że trenowane są ostatnie warstwy splotu. Pozostałe warstwy są zamrożone (nietrenowane), więc ich wagi pozostają stałe. Niższe warstwy odnoszą się do ogólnej charakterystyki. Natomiast wyższe warstwy rozpoznają bardziej złożone cechy. Pozostawienie zamrożonych warstw jest dodatkowym zabezpieczeniem przed przetrenowaniem.
