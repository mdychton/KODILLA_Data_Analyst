```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
#import danych
```

```python
dataset = pd.read_csv('titanic.csv')
```

```python
#usuniecie pustych danych
```

```python
fare_values = dataset.loc[
    dataset['Fare'].notnull(),
    'Fare'
].values

print(f'Liczba obserwacji: {fare_values.shape[0]}')
```

```python
#histogram
```

```python
plt.hist(fare_values, bins=20)

plt.title('Rozkład opłat za bilet (Fare)')
plt.xlabel('Fare')
plt.ylabel('Liczba pasażerów')

plt.show()
```

```python
#srednia
```

```python
mean_fare = np.mean(fare_values)

print(f'Średnia Fare: {mean_fare:.2f}')
```

```python
#mediana
```

```python
median_fare = np.median(fare_values)
print (f'Mediana Fare:{median_fare:2f}')
```

```python
#moda
```

```python
mode_fare = stats.mode(fare_values)

print(f'Moda Fare: {mode_fare.mode:.2f}')
print(f'Liczba wystąpień mody: {mode_fare.count}')
```

```python
#kwartyle
```

```python
q0 = np.quantile(fare_values, 0.0)
q1 = np.quantile(fare_values, 0.25)
q2 = np.quantile(fare_values, 0.50)
q3 = np.quantile(fare_values, 0.75)
q4 = np.quantile(fare_values, 1.0)

print(f'Q0 (min): {q0:.2f}')
print(f'Q1: {q1:.2f}')
print(f'Q2 (mediana): {q2:.2f}')
print(f'Q3: {q3:.2f}')
print(f'Q4 (max): {q4:.2f}')
```

```python
#histogram z liniami
```

```python

plt.hist(fare_values, bins=20)

plt.axvline(x=mean_fare, color='red', label='Średnia')
plt.axvline(x=median_fare, color='green', label='Mediana')
plt.axvline(x=mode_fare.mode, color='orange', label='Moda')

plt.axvline(x=q1, color='black', linestyle='--', label='Q1')
plt.axvline(x=q3, color='black', linestyle='--', label='Q3')

plt.title('Fare - histogram')
plt.xlabel('Fare')
plt.ylabel('Liczba pasażerów')

plt.legend()

plt.show()
```

```python
#rozstep i iqr
```

```python
fare_range = max(fare_values) - min(fare_values)

iqr = q3 - q1

print(f'Zakres: {fare_range:.2f}')
print(f'IQR: {iqr:.2f}')
```

```python
#wariancja
```

```python
variance_fare = np.var(fare_values, ddof=1)

print(f'Wariancja: {variance_fare:.2f}')
```

```python
#odchylanie standardowe
```

```python
std_fare = np.std(fare_values, ddof=1)

print(f'Odchylenie standardowe: {std_fare:.2f}')
```

```python
#boxPlot
```

```python
plt.boxplot(fare_values)

plt.title('Boxplot zmiennej Fare')

plt.ylabel('Fare')

plt.show()
```

```python
#skosnosc
```

```python
skewness = stats.skew(fare_values)

print(f'Skośność: {skewness:.2f}')
```
