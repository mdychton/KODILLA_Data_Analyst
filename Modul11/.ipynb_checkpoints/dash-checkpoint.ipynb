```python
from jupyter_dash import JupyterDash
from dash import dcc, html
from dash.dependencies import Input, Output
import plotly.graph_objects as go

# =========================
# DANE
# =========================

x_values = [1, 2, 3, 4]
y_values = [10, 20, 30, 40]

# =========================
# APLIKACJA
# =========================

app = JupyterDash(__name__)

# =========================
# LAYOUT
# =========================

app.layout = html.Div(children=[

    # Nagłówek
    html.H1(
        'Mój dashboard',
        style={'margin': 'auto'}
    ),

    # Tekst
    html.Div(
        'Tu pojawi się content',
        style={'color': 'red'}
    ),

    # Wykres
    dcc.Graph(id='first-graph'),

    # Opis slidera
    html.Label('Slider'),

    # Slider
    html.Div([
        dcc.Slider(
            id='my-slider',
            min=1,
            max=4,
            step=1,
            value=1,
            marks={
                1: '1',
                2: '2',
                3: '3',
                4: '4'
            }
        )
    ],
    style={'width': '400px'})
])

# =========================
# CALLBACK
# =========================

@app.callback(
    Output('first-graph', 'figure'),
    Input('my-slider', 'value')
)
def update_graph(slider_value):

    # bierzemy tylko część danych
    x = x_values[:slider_value]
    y = y_values[:slider_value]

    # tworzenie wykresu
    fig = go.Figure(
        data=[
            go.Scatter(
                x=x,
                y=y,
                mode='lines+markers'
            )
        ]
    )

    # layout wykresu
    fig.update_layout(
        title='Pierwszy wykres',
        width=600,
        height=600
    )

    return fig

# =========================
# START
# =========================

app.run(mode='inline')
```

Funkcja update przyjmie jeden argument: max_range, który będzie odpowiadał za szerokość osi x na wykresie. Zanim jednak uzupełnimy funkcję, omówimy sobie działanie dekoratora @app.callback.

Pierwszym argumentem dekoratora app.callback jest funkcja Output. Podajemy do niej kolejne dwa argumenty:

Id komponentu, do którego ma trafić wartość zwrócona poprzez funkcję update_figure.
component_property, czyli atrybut komponentu wskazanego przez Id, który ma być uzupełniony przez wartość funkcji update_figure.
W naszym przypadku chcemy, aby funkcja pozwalała na manipulacje wykresem, który umieściliśmy w komponencie dcc.Graph o id ‘first-graph’. Dalej, chcemy, aby wartość funkcji update_figure trafiała bezpośrednio do atrybutu figure, który do tej pory uzupełniliśmy o zmienną fig.

Kolejnym argumentem dekoratora jest lista funkcji Input. W naszym przypadku Input jest jeden (wartość slidera), dlatego lista zawiera jeden element. Argumenty funkcji Input działają tak samo, jak argumenty funkcji Output. W naszym przypadku chcemy pobierać informacje z komponentu o id ‘my-slider’, a konkretnie z jego atrybutu o nazwie ‘value’.

Możemy skopiować część kodu odpowiedzialną do tej pory za stworzenie wykresu i wkleić ją wewnątrz funkcji update_figure

Dokonujemy przy okazji jednej zmiany — tworząc layout wykresu, dodajemy parametr:

xaxis=dict(range=[1,max_range])
Uzależniamy w ten sposób maksymalną szerokość osi x od wartości slidera.

Ostatnią zmianą, jakiej musimy dokonać, jest usunięcie z layoutu aplikacji parametru figure, przy komponencie dcc.Graph o id ‘first-graph’. Od teraz ten argument będzie uzupełniany przez stworzony przez nas callback.

Całość kodu powinna wyglądać następująco:

```python
# =========================
# IMPORT BIBLIOTEK
# =========================

import dash

# Nowy sposób importu komponentów Dash
from dash import dcc, html

# Import Input i Output do callbacków
from dash.dependencies import Input, Output

# Biblioteka do tworzenia wykresów Plotly
import plotly.graph_objs as go


# =========================
# TWORZENIE APLIKACJI
# =========================

# __name__ pomaga Dash poprawnie znaleźć zasoby aplikacji
app = dash.Dash(__name__)


# =========================
# UKŁAD DASHBOARDU (LAYOUT)
# =========================

app.layout = html.Div(children=[

    # Nagłówek dashboardu
    html.H1(
        'Mój dashboard',
        style={'margin': 'auto'}
    ),

    # Krótki tekst pod nagłówkiem
    html.Div(
        'Tu pojawi się content',
        style={'color': 'red'}
    ),

    # Miejsce na wykres
    dcc.Graph(
        id='first-graph'
    ),

    # Opis slidera
    html.Label('Slider'),

    # Kontener slidera
    html.Div([

        # Slider sterujący osią X wykresu
        dcc.Slider(
            id='my-slider',

            # minimalna wartość slidera
            min=1,

            # maksymalna wartość slidera
            max=4,

            # krok zmiany
            step=1,

            # wartość początkowa
            value=1,

            # podpisy pod sliderem
            marks={
                1: '1',
                2: '2',
                3: '3',
                4: '4'
            }
        )

    ],
    
    # szerokość kontenera slidera
    style={'width': '400px'})
])


# =========================
# CALLBACK
# =========================
# Callback reaguje na zmianę slidera
# i aktualizuje wykres

@app.callback(

    # Co aktualizujemy
    Output(
        component_id='first-graph',
        component_property='figure'
    ),

    # Co wywołuje aktualizację
    Input(
        'my-slider',
        'value'
    )
)

def update_figure(max_range):

    # =========================
    # DANE DO WYKRESU
    # =========================

    data = [

        go.Scatter(

            # dane osi X
            x=[1, 2, 3, 4],

            # dane osi Y
            y=[10, 20, 30, 40],

            # typ wykresu
            mode='lines+markers'
        )
    ]

    # =========================
    # USTAWIENIA WYKRESU
    # =========================

    layout = go.Layout(

        # tytuł wykresu
        title='Pierwszy wykres',

        # szerokość wykresu
        width=600,

        # wysokość wykresu
        height=300,

        # ustawienia osi X
        xaxis=dict(

            # slider zmienia zakres osi X
            range=[1, max_range]
        )
    )

    # Tworzenie figury
    fig = go.Figure(
        data=data,
        layout=layout
    )

    # Zwracamy gotowy wykres
    return fig


# =========================
# URUCHOMIENIE APLIKACJI
# =========================

if __name__ == '__main__':

    # debug=True pokazuje błędy i automatycznie odświeża aplikację
    app.run(debug=True)
```

# Łączymy Pandas i Dash

Nadszedł czas, aby połączyć dashboard z rzeczywistą bazą danych! W tym przykładzie wykorzystamy heatmapy temperatury oraz opadów we Wrocławiu, które stworzyliśmy w części poświęconej Plotly.

W nowym dashboardzie zwiążemy informacje wyświetlane na wykresie z komponentem RangeSlider. Sprawimy, aby wykresy wyświetlały heatmapy wyłącznie dla zakresu dni wybranego przez użytkownika. Co więcej, pod wykresami będziemy na bieżąco wyświetlać informacje na temat temperatury oraz opadów dla dnia, oraz godziny, na którą użytkownik akurat najechał myszką.



```python
import dash

# Nowy sposób importu komponentów Dash
from dash import dcc, html

# Import Input i Output do callbacków
from dash.dependencies import Input, Output

# Biblioteka do tworzenia wykresów Plotly
import plotly.graph_objs as go

import pandas as pd
from plotly import tools
import datetime as dt
```

Funkcja tools z pakietu plotly posłuży nam do stworzenia dwóch wykresów na jednym płótnie, biblioteki datetime będziemy potrzebować do odpowiedniego filtrowania danych.

```python
# =========================
# IMPORT BIBLIOTEK
# =========================

import dash

# Komponenty Dash
from dash import dcc, html

# Callbacki Dash
from dash.dependencies import Input, Output

# Wykresy Plotly
import plotly.graph_objs as go

# Nowy sposób tworzenia subplotów
from plotly.subplots import make_subplots

# Operacje na datach
import datetime as dt

# Pandas
import pandas as pd


df = pd.read_csv('dataexport_20200613T163949.csv',
    skiprows=9,index_col=0,parse_dates=True)

# =========================
# TWORZENIE APLIKACJI
# =========================

app = dash.Dash(__name__)


# =========================
# LAYOUT DASHBOARDU
# =========================

app.layout = html.Div(

    children=[

        # =========================
        # TYTUŁ
        # =========================

        html.H1(
            'Pogoda we Wrocławiu',
            style={
                'text-align': 'center'
            }
        ),

        # =========================
        # WYKRES
        # =========================

        dcc.Graph(
            id='wro-weather'
        ),

        # =========================
        # RANGE SLIDER
        # =========================

        dcc.RangeSlider(

            # ID slidera
            id='date-slider',

            # minimalny dzień
            min=df.index.min().day,

            # maksymalny dzień
            max=df.index.max().day,

            # krok slidera
            step=1,

            # podpisy pod sliderem
            marks={
                date.day: date.strftime('%m-%d')
                for date in df.index.unique()
            },

            # domyślny zakres
            value=[
                df.index.min().day,
                df.index.max().day
            ]
        ),

        # =========================
        # INFORMACJE HOVER
        # =========================

        html.Div(

            id='hover-details',

            style={
                'text-align': 'center',
                'margin-top': '20px',
                'font-weight': 'bold'
            }
        )

    ],

    # Styl całego dashboardu
    style={
        'width': '80%',
        'margin': 'auto'
    }
)


# =========================
# CALLBACK - AKTUALIZACJA WYKRESU
# =========================

@app.callback(

    # Co aktualizujemy
    Output('wro-weather', 'figure'),

    # Co wywołuje aktualizację
    Input('date-slider', 'value')
)

def update_heatmap(x_range):

    # =========================
    # FILTROWANIE DANYCH
    # =========================

    wro_temp = df.truncate(

        before=dt.datetime(2020, 6, x_range[0]),

        after=dt.datetime(2020, 6, x_range[1]) +
        pd.Timedelta(hours=23)
    )

    # =========================
    # TWORZENIE SUBPLOTÓW
    # =========================

    fig = make_subplots(

        rows=1,
        cols=2,

        subplot_titles=[
            'Temperatura',
            'Opady'
        ],

        shared_yaxes=True
    )

    # =========================
    # MAPA DNI TYGODNIA
    # =========================

    weekday_map = {

        0: 'Poniedziałek',
        1: 'Wtorek',
        2: 'Środa',
        3: 'Czwartek',
        4: 'Piątek',
        5: 'Sobota',
        6: 'Niedziela'
    }

    # =========================
    # HEATMAPA TEMPERATURY
    # =========================

    fig.add_trace(

        go.Heatmap(

            x=wro_temp.index.hour,

            y=wro_temp.index.weekday.map(
                weekday_map
            ),

            z=wro_temp[
                'Wrocław Temperature [2 m elevation corrected]'
            ].tolist(),

            colorscale='Jet',

            showscale=False
        ),

        row=1,
        col=1
    )

    # =========================
    # HEATMAPA OPADÓW
    # =========================

    fig.add_trace(

        go.Heatmap(

            x=wro_temp.index.hour,

            y=wro_temp.index.weekday.map(
                weekday_map
            ),

            z=wro_temp[
                'Wrocław Precipitation Total'
            ].tolist(),

            colorscale='Cividis',

            showscale=False
        ),

        row=1,
        col=2
    )

    # =========================
    # USTAWIENIA LAYOUTU
    # =========================

    fig.update_layout(

        title='Analiza pogody',

        height=500,

        margin=dict(
            l=40,
            r=40,
            t=80,
            b=40
        )
    )

    return fig


# =========================
# CALLBACK - HOVER
# =========================

@app.callback(

    Output('hover-details', 'children'),

    Input('wro-weather', 'hoverData')
)

def update_hover_details(hoverData):

    # Jeśli użytkownik jeszcze nie najechał
    if hoverData is None:

        return 'Najedź myszką na wykres'

    # Pobranie punktu hover
    dpoint = hoverData['points'][0]

    # =========================
    # TEMPERATURA
    # =========================

    if dpoint['curveNumber'] == 0:

        return (
            f"{dpoint['y']}: "
            f"temperatura o godzinie "
            f"{dpoint['x']} wynosiła "
            f"{round(dpoint['z'], 2)}°C"
        )

    # =========================
    # OPADY
    # =========================

    elif dpoint['curveNumber'] == 1:

        return (
            f"{dpoint['y']}: "
            f"opady o godzinie "
            f"{dpoint['x']} wynosiły "
            f"{round(dpoint['z'], 2)} mm"
        )


# =========================
# START APLIKACJI
# =========================

if __name__ == '__main__':

    app.run(debug=True)
```

# Dash State 

Do tej pory używaliśmy callbacków do natychmiastowego odświeżania danych widocznych na dashboardzie. Heatmapy aktualizowaliśmy zaraz po wykonaniu jakiegokolwiek ruchu na sliderze, dane znajdujące się w miejscu kursora wyświetlaliśmy na nowo z każdym ruchem myszki.

Nie zawsze jednak będzie to pożądane zachowanie aplikacji — w niektórych przypadkach będzie nam zależeć na tym, aby dashboard odświeżał się po uruchomieniu konkretnego wyzwalacza.

Weźmy następujący przykład. Nasz dashboard zawiera dwa komponenty typu Input. Działanie dashboardu polega na przechwytywaniu wartości z tych komponentów, a następnie zwracaniu ich iloczynu do stworzonego w tym celu miejsca.

Kod źródłowy tego prostego dashboardu będzie wyglądał następująco:

```python
# =========================
# IMPORT BIBLIOTEK
# =========================

import dash

# Nowy sposób importu komponentów Dash (zamiast dash_core_components)
from dash import dcc, html

# Callbacki (Input/Output)
from dash.dependencies import Input, Output


# =========================
# TWORZENIE APLIKACJI
# =========================

app = dash.Dash(__name__)


# =========================
# LAYOUT (INTERFEJS)
# =========================

app.layout = html.Div([

    # Tytuł aplikacji
    html.H1('Kalkulator'),

    # =========================
    # INPUT 1
    # =========================

    html.Label('Czynnik #1'),

    dcc.Input(
        id='product-1',          # ID komponentu
        placeholder='5',         # tekst podpowiedzi
        type='number',           # tylko liczby
        value=1                  # wartość domyślna (ważne: int, nie string)
    ),

    html.Br(),

    # =========================
    # INPUT 2
    # =========================

    html.Label('Czynnik #2'),

    dcc.Input(
        id='product-2',
        placeholder='2',
        type='number',
        value=1
    ),

    # =========================
    # WYNIK
    # =========================

    html.Div(
        id='output',
        style={
            'font-size': '30px',
            'margin': '10px 10px'
        }
    )
])


# =========================
# CALLBACK (LOGIKA)
# =========================
# Tu dodasz mnożenie / obliczenia później

@app.callback(
    Output('output', 'children'),
    [
        Input('product-1', 'value'),
        Input('product-2', 'value')
    ]
)

def calculate(p1, p2):

    # zabezpieczenie przed None
    if p1 is None or p2 is None:
        return "Wpisz obie liczby"

    # wynik mnożenia
    result = p1 * p2

    return f"Wynik: {result}"


# =========================
# URUCHOMIENIE
# =========================

if __name__ == '__main__':
    app.run(debug=True)
```

Tworzymy zatem layout z dwoma komponentami typu Input, którym nadajemy id odpowiednio 'product-1' oraz 'product-2'. Wynik mnożenia liczb z tych pól zwrócimy w miejsce oznaczone przez id 'output'.

Jeśli jednak będziemy chcieli wykonać inne mnożenie i wrócimy do pola product-1, zmieniając jego wartość, wartość widoczna w output stanie się wynikiem mnożenia nowej wartości w product-1 oraz starej wartości product-2. Na potrzeby przykładu uznajemy, że jest to działanie niepożądane.

Będziemy chcieli poprawić funkcjonalność dashboardu w taki sposób, aby mnożenie wykonywało się dopiero po zatwierdzeniu przez użytkownika. W tym celu pod polami z inputem dodamy przycisk wyzwalający funkcję multiply.

W związku z tym zmienimy layout naszego dashboardu. Pod komponentem o id 'product-2' dodamy Button i nadamy mu id 'submit-btn'.

Zwróćmy uwagę, że Button należy do dash_html_components, a nie do dash_core_components.

Dashboard w wersji 2.0, zamiast przechwytywać dane z komponentów Input i uruchamiać od razu funkcję multiply, będzie musiał przechowywać dane do momentu uruchomienia wyzwalacza. Za przechowywanie danych będzie odpowiadać funkcja State.

Funkcję State importujemy w jednej linii razem z funkcjami Input oraz Output.

Uzupełniamy callback o trzeci argument — listę funkcji State. Inputem w tej wersji staje się Button, który dodaliśmy wcześniej, a atrybut, który pobieramy to n_clicks.

Atrybut n_clicks zlicza, ile razy użytkownik kliknął w dany komponent od momentu uruchomienia dashboardu.

Z kolei funkcje State przejmują zadanie przechowywania wartości z komponentów product-1 oraz product-2 do momentu kliknięcia przez użytkownika w element wskazany w funkcji Input.

Teraz dashboard działa zgodnie z podstawowymi założeniami:

```python
# =========================
# IMPORTY
# =========================

import dash
from dash import dcc, html
from dash.dependencies import Input, Output, State



# =========================
# APLIKACJA
# =========================

app = dash.Dash(__name__)


# =========================
# LAYOUT (INTERFEJS)
# =========================

app.layout = html.Div([

    # Tytuł
    html.H1('Kalkulator'),

    # =========================
    # INPUT 1
    # =========================

    html.Label('Czynnik #1'),

    dcc.Input(
        id='product-1',
        type='number',
        value=1,
        placeholder='5'
    ),

    html.Br(),

    # =========================
    # INPUT 2
    # =========================

    html.Label('Czynnik #2'),

    dcc.Input(
        id='product-2',
        type='number',
        value=1,
        placeholder='2'
    ),

    html.Br(),
    html.Br(),

    # =========================
    # PRZYCISK (TRIGGER CALLBACKA)
    # =========================

    html.Button(
        'Oblicz',
        id='submit-btn',
        n_clicks=0   # licznik kliknięć
    ),

    html.Br(),
    html.Br(),

    # =========================
    # OUTPUT
    # =========================

    html.Div(
        id='output',
        style={
            'font-size': '30px',
            'margin': '10px 10px'
        }
    )
])


# =========================
# CALLBACK (LOGIKA)
# =========================

@app.callback(

    # Output: gdzie zapisujemy wynik
    Output('output', 'children'),

    # Input: kliknięcie przycisku uruchamia callback
    Input('submit-btn', 'n_clicks'),

    # State: pobiera aktualne wartości inputów w momencie kliknięcia
    [
        State('product-1', 'value'),
        State('product-2', 'value')
    ]
)

def multiply(n_clicks, prod_1, prod_2):

    # =========================
    # ZABEZPIECZENIE
    # =========================

    if n_clicks is None or n_clicks == 0:
        return "Kliknij 'Oblicz'"

    if prod_1 is None or prod_2 is None:
        return "Wpisz obie liczby"

    # =========================
    # OBLICZENIE
    # =========================

    result = prod_1 * prod_2

    return f"Wynik: {result}"


# =========================
# URUCHOMIENIE APLIKACJI
# =========================

if __name__ == '__main__':
    app.run(debug=True)
```
